### ✏️ [문제 1] print() 심화

아래 출력 예시가 **정확히** 나오도록 `print()` 문을 작성하세요.
각 줄에 사용해야 할 힌트를 잘 읽어보세요.

```
==== 입원 환자 정보 ====
환자명 : 박지성 | 나이 : 38세
혈압   : 125 / 82 mmHg
체온   : 37.4 ℃  →  정상
========================
```

- 첫째·마지막 줄: `=` 24개
- `|` 구분자는 `sep` 없이 f-string으로 표현하세요
- `→` 는 `\u2192` 유니코드 문자입니다


````python
# 변수 먼저 선언하세요
name = "박지성"
age  = 38
sbp, dbp = 125, 82   # 수축기/이완기 혈압
temp = 37.4

# 코드 작성
print("==== 입원 환자 정보 ====")
print(f"환자명 : {name} | 나이 : {age}세")
print(f"혈압   : {sbp} / {dbp} mmHg")
print(f"체온   : {temp} ℃  →  정상")
print("========================")
````

### ✏️ [문제 2] 변수 선언 + 복합 계산

아래 정보를 변수로 저장한 뒤 **두 가지 계산 결과**를 f-string으로 출력하세요.

- 환자명: `"강하늘"` / 나이: `29` / 키: `175.8`cm / 체중: `82.3`kg
- 계산 ①: BMI = 체중(kg) ÷ (키(m))² → 소수점 1자리
- 계산 ②: 이상 체중 = 키(cm) - 100 → 이상 체중 대비 현재 체중 차이(±)

출력 예시:
```
강하늘(29세) | BMI: 26.6 | 이상체중 대비: +6.3kg
```


````python
# 코드 작성
patient_name = "강하늘"
patient_age  = 29
patient_height = 175.8
patient_weight = 82.3

print(f"""{patient_name}({patient_age}세) | BMI: {patient_weight / (patient_height / 100) ** 2:.1f}
  | 이상체중 대비: {int((patient_weight - (patient_height - 100)) * 1000) / 1000:.1f}""")
print(f"이상체중 대비: {patient_weight - (patient_height - 100)}") #이상체중 대비: 6.499999999999986 -> 절대로 6.3kg이 나올 수 없음
````

### ✏️ [문제 11] 처방 계산 심화 🚀

아래 조건으로 항생제 처방 계산을 완성하세요.

- 환자 체중: `63`kg / kg당 용량: `7.5`mg / 하루 2회 / 5일 처방
- 병원 보유 앰플: 1앰플 = 250mg
- 1회 필요 용량 (소수점 1자리)
- 5일 총 필요량 (소수점 없음)
- 필요 앰플 수: 총량 ÷ 250 → **올림** (힌트: `import math; math.ceil()`)
- 남은 약 = 앰플 수 × 250 − 총 필요량

출력 예시:
```
1회 용량:    236.2mg
5일 총량:    2362mg
필요 앰플:   10개
남는 양:     138mg
```


````python
import math

weight = 63
dose_per_kg = 7.5
freq = 2
days = 5
amp_size = 250   # 1앰플 용량(mg)

# 계산
one_dose = weight * dose_per_kg / freq
total_dose = weight * dose_per_kg * days
ampules = math.ceil(total_dose / amp_size)
remain = ampules * amp_size - total_dose

# 출력
print(f"1회 용량:    {one_dose:.1f}mg")
print(f"5일 총량:    {total_dose:.0f}mg")
print(f"필요 앰플:   {ampules}개")
print(f"남는 양:     {remain:.0f}mg")
````

### ✏️ [문제 12] 복합 대입 연산자 + 누적 통계

`+=` / `-=` / `//=` 를 사용해 아래 재고를 관리하고 통계를 출력하세요.

- 초기 재고: `1000`정
- 월요일 출고: `230`정 → 재고 출력
- 화요일 입고: `500`정 → 재고 출력
- 수요일 출고: `315`정 → 재고 출력
- 목요일 출고: `180`정 → 재고 출력
- 평균 일일 출고량 = 총 출고량 ÷ 출고 일수 (정수 나눗셈)

출력 예시:
```
월요일 재고: 770정
화요일 재고: 1270정
수요일 재고: 955정
목요일 재고: 775정
총 출고량: 725정 | 평균 일일 출고: 241정
```


````python
stock = 1000
total_out = 0

# 월요일 출고
stock -= 230
total_out += 230
print(f"월요일 재고: {stock}정")

# 화요일 입고
stock += 500
print(f"화요일 재고: {stock}정")

# 수요일 출고
stock -= 315
total_out += 315
print(f"수요일 재고: {stock}정")

# 목요일 출고
stock -= 180
total_out += 180
print(f"목요일 재고: {stock}정")

# 평균 출고량
average_out = total_out // 3

print(f"총 출고량: {total_out}정 | 평균 일일 출고: {average_out}정")
````

## ❓ 종합 문제 5: 응급 중증도 종합 판정 시스템 🚀

오늘 배운 모든 개념을 활용해 KTAS(한국형 응급환자 분류) 간이 판정기를 만드세요.

**환자 정보:**
- 이름: `"오준혁"` / 나이: `61`
- 수축기혈압: `90` / 심박수: `122` / 체온: `38.9` / SpO2: `92`
- 의식: `"혼미"` (정상=`"명료"`) / 주증상: `"흉통"`

**판정 기준:**
- 즉시(1등급): 의식 비정상 OR SpO2 < 90
- 긴급(2등급): 즉시 아니고, (혈압<100 OR 맥박>120 OR 체온>38.5)
- 응급(3등급): 1·2등급 아니고, (나이≥60 AND 체온>37.5)
- 고위험군: 나이≥60 AND (고혈압기준 OR 빈맥기준)
  - 고혈압기준: 혈압 < 100
  - 빈맥기준: 맥박 > 100

출력 예시:
```
===== 오준혁(61세) KTAS 판정 =====
즉시(1등급): True
긴급(2등급): False
응급(3등급): False
고위험군:    True
주증상: 흉통 | 의식: 혼미
```


````python
# 환자 정보
name = "오준혁"
age = 61
systolic_bp = 90 # 수축기 혈압
heart_rate = 122 # 심박수
temperature = 38.9 # 체온
spo2 = 92 # 산소포화도
mental = "혼미" # 의식 상태
symptom = "흉통" # 주증상
# ========================= # KTAS 판정 # =========================
# 1등급: 즉시
immediate = (mental != "명료") or (spo2 < 90)
# 2등급: 긴급
urgent = (not immediate) and ( (systolic_bp < 100) or (heart_rate > 120) or (temperature > 38.5) )
# 3등급: 응급
emergency = (not immediate) and (not urgent) and ( (age >= 60) and (temperature > 37.5) )
# 고위험군 판정
high_bp_risk = systolic_bp < 100
tachycardia_risk = heart_rate > 100
high_risk = (age >= 60) and ( high_bp_risk or tachycardia_risk )
# ========================= # 결과 출력 # =========================
print(f"===== {name}({age}세) KTAS 판정 =====")
print(f"즉시(1등급): {immediate}")
print(f"긴급(2등급): {urgent}")
print(f"응급(3등급): {emergency}")
print(f"고위험군: {high_risk}")
print(f"주증상: {symptom} | 의식: {mental}")
````
