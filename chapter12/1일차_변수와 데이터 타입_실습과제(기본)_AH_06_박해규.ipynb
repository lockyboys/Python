# 실습과제: 환자 기본정보 저장 및 계산하기

환자의 기본정보를 변수에 저장하고, 다양한 연산자를 활용하여 의료 정보를 계산하는 프로그램을 작성해보세요.

### 요구사항

1. **환자 기본정보 저장**
   - 환자 이름: "박건강"
   - 출생연도: 1990년
   - 키: 170cm  
   - 몸무게: 65kg
   - 성별: "남성"

2. **계산 작업**
   - BMI 계산 (공식: 몸무게(kg) ÷ 키(m)²)
   - 현재 나이 계산 (현재 연도: current_year)
   - 성인 여부 판별 (18세 이상)

3. **출력 형식**
```
=== 환자 정보 ===
이름: 박건강 (남성, 35세)
키: 170cm, 몸무게: 65kg
BMI: 22.5
출생연도: 1990년
성인 여부: True
```

### 힌트
- BMI 계산 시 키를 cm에서 m로 변환해야 합니다
- 현재 연도에서 출생연도를 빼면 나이가 나옵니다
- 논리 연산자로 성인 여부를 확인하세요
- f-string으로 깔끔하게 출력하세요
