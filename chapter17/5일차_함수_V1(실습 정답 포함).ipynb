#  1. 함수란? — 왜 필요한가?

## 함수(Function)가 없으면 어떻게 불편할까요?

> 💡 **비유:** 함수는 **레시피 카드**입니다.
> 떡볶이 만드는 법을 매번 처음부터 생각하는 대신,
> "떡볶이 만들기" 레시피 카드를 보고 그대로 하면 됩니다.
> 함수도 마찬가지 — 코드를 한 번 작성해두고 이름만 불러서 재사용합니다.


### 😓 함수 없이 — 같은 코드를 반복 작성해야 합니다

````python
# 환자 3명의 BMI를 계산하려면?
# 같은 계산식을 3번 반복 작성해야 합니다

# 환자 1
h1, w1 = 170.0, 75.0
bmi1   = w1 / ((h1 / 100) ** 2)
print(f"김환자 BMI: {bmi1:.2f}")

# 환자 2
h2, w2 = 160.5, 58.2
bmi2   = w2 / ((h2 / 100) ** 2)
print(f"이환자 BMI: {bmi2:.2f}")

# 환자 3
h3, w3 = 180.0, 85.0
bmi3   = w3 / ((h3 / 100) ** 2)
print(f"박환자 BMI: {bmi3:.2f}")

# ⚠️ 공식이 바뀌면? 3군데 모두 수정해야 합니다!
````

### 😊 함수 사용 — 한 번 정의하고 이름만 불러서 재사용합니다

````python
# 함수를 한 번 정의하면
def calculate_bmi(height_cm, weight_kg):
    # calculate(계산하다) + bmi(체질량지수)
    # height_cm(키, cm) / weight_kg(체중, kg)
    bmi = weight_kg / ((height_cm / 100) ** 2)
    return round(bmi, 2)

# 이름만 불러서 몇 번이든 재사용!
print(f"김환자 BMI: {calculate_bmi(170.0, 75.0)}")
print(f"이환자 BMI: {calculate_bmi(160.5, 58.2)}")
print(f"박환자 BMI: {calculate_bmi(180.0, 85.0)}")

# ✅ 공식이 바뀌면? 함수 안 한 곳만 수정하면 끝!
````

## 함수의 3가지 핵심 장점

| 장점 | 설명 | 비유 |
|------|------|------|
| **재사용** | 한 번 만들면 여러 곳에서 반복 사용 | 레시피 카드 |
| **유지보수** | 수정이 필요하면 함수 안 한 곳만 고침 | 원본 문서 수정 |
| **가독성** | `calculate_bmi(170, 70)` — 코드가 말처럼 읽힘 | 목차 있는 책 |

#  2. 함수의 기본 구조

```
def  함수이름  (매개변수1, 매개변수2, ...):
 ↑      ↑           ↑
정의   이름표    입력값 자리
    실행할 코드
    return 반환값   ← 결과를 돌려줌
```

> ⚠️ **규칙 2가지**
> 1. `def` 뒤에 **콜론(:)** 필수
> 2. 함수 안의 코드는 **4칸 들여쓰기** (3일차 조건문, 4일차 반복문과 동일)


````python
# 가장 간단한 함수 — 매개변수도 반환값도 없는 형태
def greet_patient():
    # greet(인사하다) + patient(환자)
    print("안녕하세요, 환자님!")
    print("오늘 진료에 오신 것을 환영합니다.")

# 함수 정의만 해서는 아무것도 실행되지 않습니다
# 반드시 '호출'해야 실행됩니다

greet_patient()   # 함수 호출 — 이름 뒤에 ()를 붙입니다
greet_patient()   # 몇 번이든 재사용 가능
greet_patient()
````

````python
# 함수 정의 ≠ 함수 실행 — 차이를 확인해봐요
print("1. 이 줄이 먼저 출력됩니다")

def say_hello():
    print("3. 함수 안의 코드 — 호출될 때만 실행!")

print("2. 함수를 정의했지만 아직 실행 안 됨")
say_hello()     # 여기서 비로소 함수 안 코드 실행
print("4. 함수 호출 후 계속 진행")
````

#  3. 매개변수(Parameter)와 인수(Argument)

함수가 정말 유용해지려면 **외부에서 값을 받아서** 처리해야 합니다.
그 통로가 바로 매개변수(parameter)와 인수(argument)입니다.

> 💡 **비유로 이해하기**
>
> ```
> 함수 정의:  def 커피_만들기(원두_종류, 설탕_개수):
>                              ↑           ↑
>                          매개변수(parameter) — 빈 컵 자리
>
> 함수 호출:  커피_만들기("에티오피아", 2)
>                          ↑             ↑
>                       인수(argument) — 실제로 넣는 재료
> ```
>
> | 용어 | 위치 | 역할 |
> |------|------|------|
> | **매개변수(parameter)** | 함수 **정의**할 때 | 값을 받을 빈 자리 이름 |
> | **인수(argument)** | 함수 **호출**할 때 | 실제로 전달하는 값 |


````python
# 매개변수가 있는 함수
def register_patient(name, age, department):
    # name(이름), age(나이), department(진료과)
    # ↑ 이 셋이 매개변수 — 함수가 받을 값의 자리

    print(f"환자 등록 완료!")
    print(f"  이름: {name}")
    print(f"  나이: {age}세")
    print(f"  진료과: {department}")

# 함수 호출 — 인수(실제 값)를 전달
register_patient("김환자", 45, "내과")
#                  ↑       ↑     ↑
#         인수들 — name, age, department 에 순서대로 들어감

print()
register_patient("이환자", 32, "정형외과")
````

````python
# 변수를 인수로 전달하는 경우
patient_name  = "박환자"   # patient_name(환자이름)
patient_age   = 58         # patient_age(환자나이)
patient_dept  = "내분비내과" # patient_dept(진료과)

register_patient(patient_name, patient_age, patient_dept)
# 변수의 값이 매개변수로 전달됩니다
````

````python
# 매개변수와 인수 개수가 다르면 오류!
def bmi_calculator(height_cm, weight_kg):
    # height_cm(키), weight_kg(체중)
    return weight_kg / ((height_cm / 100) ** 2)

# ❌ 인수 1개 — 매개변수 2개인데 1개만 전달
# bmi_calculator(170)   # TypeError 발생!

# ✅ 인수 2개 — 정확히 일치
result = bmi_calculator(170, 70)
print(f"BMI: {result:.2f}")

# ✅ 변수로 전달도 OK
h = 165.5   # h(height, 키)
w = 58.2    # w(weight, 체중)
print(f"BMI: {bmi_calculator(h, w):.2f}")
````

#  4. 반환값 — return

`return`은 함수가 계산한 결과를 **호출한 곳으로 돌려주는** 명령입니다.

> 💡 **비유:** 함수를 자판기라고 생각해봐요.
> 돈(인수)을 넣으면 → 음료(반환값)가 나옵니다.
> `return`이 없으면 돈만 먹고 아무것도 안 주는 자판기!

```python
# return 없는 함수 — 실행만 하고 끝
def print_bmi(bmi):
    print(f"BMI: {bmi}")   # 출력만, 값을 돌려주지 않음

# return 있는 함수 — 결과를 돌려줌 (다른 곳에 활용 가능)
def calculate_bmi(height, weight):
    bmi = weight / ((height / 100) ** 2)
    return bmi              # 계산 결과를 돌려줌
```


````python
# return이 없는 함수 vs return이 있는 함수 비교

# ① return 없는 함수
def print_bmi_no_return(height_cm, weight_kg):
    bmi = weight_kg / ((height_cm / 100) ** 2)
    print(f"BMI: {bmi:.2f}")      # 출력만 함

# ② return 있는 함수
def calculate_bmi(height_cm, weight_kg):
    # height_cm(키, cm), weight_kg(체중, kg)
    bmi = weight_kg / ((height_cm / 100) ** 2)
    return round(bmi, 2)          # 값을 돌려줌

# 차이 비교
result1 = print_bmi_no_return(170, 75)   # 출력만
print(f"return 없는 함수의 결과: {result1}")  # None!

result2 = calculate_bmi(170, 75)          # 값을 반환
print(f"return 있는 함수의 결과: {result2}")  # 23.18

# return이 있으면 결과를 다른 곳에 활용 가능
if result2 >= 25:
    print("과체중 — 식이요법 상담 권장")
else:
    print("정상 체중")
````

### return의 두 번째 역할 — 함수 중간에 종료

````python
# return은 값 반환 외에 '함수 즉시 종료'에도 쓰입니다
def check_patient_age(age):
    # age(나이)
    # 잘못된 입력이면 즉시 종료 (방어 코드)
    if age < 0:
        print("오류: 나이는 0 이상이어야 합니다.")
        return      # ← 값 없이 return → 함수 종료

    if age < 18:
        print(f"{age}세 → 소아 환자: 소아과 안내")
        return      # ← 여기서 종료, 아래 코드는 실행 안 됨

    # 18세 이상만 여기까지 도달
    print(f"{age}세 → 성인 환자: 일반 진료 안내")

check_patient_age(-1)
print("---")
check_patient_age(15)
print("---")
check_patient_age(45)
````

### 여러 값을 한 번에 반환하기 — 튜플 (2일차 연결)

````python
# 함수는 여러 값을 반환할 수 있습니다 → 자동으로 튜플로 묶임
def get_vital_summary(temperatures, blood_pressures):
    # temperatures(체온 리스트), blood_pressures(혈압 리스트)
    avg_temp = round(sum(temperatures) / len(temperatures), 1)
    max_bp   = max(blood_pressures)
    min_bp   = min(blood_pressures)
    return avg_temp, max_bp, min_bp     # 쉼표로 구분 → 튜플로 반환

# 반환된 값을 튜플 언패킹으로 받기 (2일차 튜플 연결)
temp_avg, bp_max, bp_min = get_vital_summary(
    [36.5, 37.2, 38.1, 36.8],   # 체온 기록
    [120, 145, 132, 158, 115]    # 혈압 기록
)

print(f"평균 체온:  {temp_avg}℃")
print(f"최고 혈압:  {bp_max}mmHg")
print(f"최저 혈압:  {bp_min}mmHg")

# 튜플 전체로 받을 수도 있음
result = get_vital_summary([36.5, 37.2, 38.1], [120, 145, 132])
print(f"
반환 타입: {type(result)}")  # <class 'tuple'>
print(f"튜플 전체: {result}")
````

````python
# 3일차 조건문 + return 결합 — 조건에 따라 다른 값 반환
def classify_blood_pressure(systolic):
    # systolic(수축기 혈압), classify(분류하다)
    # 수축기 혈압을 받아서 단계를 반환합니다
    if systolic < 120:
        return "정상"
    elif systolic < 130:
        return "높은 정상"
    elif systolic < 140:
        return "고혈압 전단계"
    elif systolic < 160:
        return "고혈압 1기"
    else:
        return "고혈압 2기"

# 반환값을 바로 변수에 저장
bp_values = [115, 125, 138, 152, 168]

print("=== 혈압 단계 분류 ===")
for bp in bp_values:
    stage = classify_blood_pressure(bp)   # stage(단계)
    print(f"  {bp}mmHg → {stage}")
````

### ✏️ [직접해보기 4-1] 산소포화도 판정 함수

````python
# 산소포화도를 받아서 상태를 반환하는 함수를 작성하세요
# 함수명: classify_spo2
# 반환 기준:
#   95 이상     → "정상"
#   90 ~ 94    → "경증 저산소증"
#   85 ~ 89    → "중등도 저산소증"
#   85 미만     → "중증 저산소증"
#
# 테스트:
# print(classify_spo2(97))  → 정상
# print(classify_spo2(92))  → 경증 저산소증

# 코드 작성

````

````python
# @title 정답 보기
def classify_spo2(spo2):
    # spo2(oxygen saturation, 산소포화도), classify(분류하다)
    if spo2 >= 95:
        return "정상"
    elif spo2 >= 90:
        return "경증 저산소증"
    elif spo2 >= 85:
        return "중등도 저산소증"
    else:
        return "중증 저산소증"

# 테스트
test_values = [97, 92, 87, 82]
for val in test_values:
    print(f"  SpO2 {val}% → {classify_spo2(val)}")
````

### ✏️ [직접해보기 4-2] 여러 값 반환 + 언패킹

````python
# 체온 리스트를 받아서 평균·최고·최저·발열 환자 수를 반환하는 함수
# 함수명: analyze_temperatures
# 반환: (평균, 최고, 최저, 발열_환자수)  ← 튜플로 반환
# 발열 기준: 37.5℃ 이상

# 코드 작성

````

````python
# @title 정답 보기
def analyze_temperatures(temps):
    # temps(temperatures, 체온 리스트)
    avg_temp   = round(sum(temps) / len(temps), 1)
    max_temp   = max(temps)
    min_temp   = min(temps)
    fever_count= sum(1 for t in temps if t >= 37.5)
    # fever_count(발열 환자 수)
    return avg_temp, max_temp, min_temp, fever_count

# 언패킹으로 받기
avg, high, low, fevers = analyze_temperatures(
    [36.5, 38.2, 37.1, 39.0, 36.8, 37.6]
)
print(f"평균 체온:    {avg}℃")
print(f"최고 체온:    {high}℃")
print(f"최저 체온:    {low}℃")
print(f"발열 환자 수: {fevers}명")
````

#  5. 기본값 매개변수 (Default Parameter)

함수를 호출할 때 특정 인수를 생략해도 되도록 **미리 기본값을 설정**하는 방법입니다.

> 💡 **비유:** 커피숍 주문 — "아이스 아메리카노 주세요"
> → 사이즈를 말 안 했지만 기본(미디엄)으로 나옵니다.
> → "라지로 주세요"라고 하면 기본값 대신 라지로 나옵니다.


````python
# 기본값 없을 때의 문제 — 항상 모든 인수를 전달해야 함
def old_schedule(patient_name, doctor, time, room):
    print(f"{patient_name} → {doctor} ({time}, {room})")

old_schedule("김환자", "이내과", "14:00", "101호")  # OK
# old_schedule("박환자")   # ❌ TypeError: 인수 3개 부족!
````

# 🗂️ 7. 자료구조와 함수 결합 (2일차 연결)

함수는 리스트·딕셔너리·집합·튜플을 **매개변수로 받거나 반환값으로 돌려줄 수** 있습니다.
2일차에서 배운 자료구조와 함수를 결합하면 훨씬 강력해집니다.


````python
# 리스트를 받아서 처리하는 함수
def filter_high_bp_patients(patients, threshold=140):
    # patients(환자 리스트), threshold(기준값)
    # filter(걸러내다), high(높은), bp(혈압)
    high_bp = [p for p in patients if p["bp"] >= threshold]
    return high_bp

# 딕셔너리 리스트 (2일차 연결)
ward = [
    {"name": "김환자", "age": 72, "bp": 158, "temp": 38.4},
    {"name": "이환자", "age": 45, "bp": 120, "temp": 36.8},
    {"name": "박환자", "age": 68, "bp": 165, "temp": 37.2},
    {"name": "최환자", "age": 38, "bp": 115, "temp": 38.9},
    {"name": "정환자", "age": 80, "bp": 142, "temp": 36.5},
]

# 함수 호출
result = filter_high_bp_patients(ward)
print(f"혈압 140 이상 환자: {len(result)}명")
for p in result:
    print(f"  {p['name']}({p['age']}세) — {p['bp']}mmHg")

print()
result2 = filter_high_bp_patients(ward, threshold=150)
print(f"혈압 150 이상 환자: {len(result2)}명")
for p in result2:
    print(f"  {p['name']} — {p['bp']}mmHg")
````

````python
# 딕셔너리를 반환하는 함수
def create_patient_chart(name, age, diagnosis, vitals_dict):
    # vitals_dict(활력징후 딕셔너리)
    # 함수 안에서 딕셔너리를 조립해서 반환
    chart = {
        "name":      name,
        "age":       age,
        "diagnosis": diagnosis,
        "vitals":    vitals_dict,       # 입력 딕셔너리 그대로 저장
        "alerts":    [],                # 이상 항목 목록 (빈 리스트로 초기화)
    }

    # 3일차 조건문으로 이상 항목 감지
    if vitals_dict.get("bp", 0) >= 140:
        chart["alerts"].append("고혈압")
    if vitals_dict.get("temp", 0) >= 37.5:
        chart["alerts"].append("발열")
    if vitals_dict.get("spo2", 100) < 95:
        chart["alerts"].append("저산소증")

    return chart

# 사용
chart = create_patient_chart(
    name      = "홍환자",
    age       = 67,
    diagnosis = "고혈압",
    vitals_dict = {"bp": 155, "temp": 38.2, "spo2": 93, "hr": 98}
)

print(f"환자: {chart['name']}({chart['age']}세)")
print(f"진단: {chart['diagnosis']}")
print(f"이상 항목: {chart['alerts'] if chart['alerts'] else '없음'}")
````

````python
# 집합을 활용한 처방 안전성 검사 함수 (2일차 집합 연결)
def check_prescription_safety(patient_allergies, new_drugs):
    # patient_allergies(환자 알레르기 집합)
    # new_drugs(새 처방 집합), check(확인)
    conflict   = patient_allergies & new_drugs    # 교집합
    safe_drugs = sorted(new_drugs - patient_allergies)  # 차집합

    return {
        "safe":     len(conflict) == 0,   # safe(안전여부)
        "conflict": conflict,              # conflict(충돌 약물)
        "safe_rx":  safe_drugs,           # safe_rx(안전한 처방)
    }

# 테스트
result = check_prescription_safety(
    patient_allergies = {"페니실린", "아스피린", "설파제"},
    new_drugs         = {"메트포르민", "아스피린", "암로디핀", "설파제"}
)

print(f"처방 안전: {result['safe']}")
if not result["safe"]:
    print(f"⚠️  충돌 약물: {result['conflict']}")
print(f"✅ 안전한 처방: {result['safe_rx']}")
````

### ✏️ [직접해보기 7] 딕셔너리 리스트 처리 함수

````python
# 환자 리스트를 받아서 통계를 딕셔너리로 반환하는 함수를 작성하세요
# 함수명: summarize_ward
# 반환: {"total": 총환자수, "avg_bp": 평균혈압, "fever_count": 발열수}

patients = [
    {"name": "김환자", "bp": 155, "temp": 38.3},
    {"name": "이환자", "bp": 120, "temp": 36.8},
    {"name": "박환자", "bp": 162, "temp": 37.9},
    {"name": "최환자", "bp": 118, "temp": 36.5},
    {"name": "정환자", "bp": 142, "temp": 37.6},
]

# 코드 작성

````

````python
# @title 정답 보기
def summarize_ward(patients):
    # summarize(요약하다), ward(병동)
    bp_list      = [p["bp"]   for p in patients]
    fever_count  = sum(1 for p in patients if p["temp"] >= 37.5)
    # fever_count(발열 환자 수)

    return {
        "total":       len(patients),
        "avg_bp":      round(sum(bp_list) / len(bp_list), 1),
        "max_bp":      max(bp_list),
        "fever_count": fever_count,
    }

patients = [
    {"name": "김환자", "bp": 155, "temp": 38.3},
    {"name": "이환자", "bp": 120, "temp": 36.8},
    {"name": "박환자", "bp": 162, "temp": 37.9},
    {"name": "최환자", "bp": 118, "temp": 36.5},
    {"name": "정환자", "bp": 142, "temp": 37.6},
]

stats = summarize_ward(patients)
print(f"총 환자:    {stats['total']}명")
print(f"평균 혈압:  {stats['avg_bp']}mmHg")
print(f"최고 혈압:  {stats['max_bp']}mmHg")
print(f"발열 환자:  {stats['fever_count']}명")
````

# ⚡ 8. 람다 함수 (Lambda Function)

람다(lambda) 함수는 **간단한 함수를 한 줄로** 만드는 방법입니다.
이름 없이 바로 쓰고 버리는 **일회용 함수**입니다.

> 💡 **비유:** 일반 함수가 레시피 카드라면,
> 람다 함수는 포스트잇에 빠르게 적은 메모입니다.
> 복잡한 요리엔 레시피 카드, 간단한 메모엔 포스트잇!

```python
# 일반 함수
def double(x):
    return x * 2

# 람다 함수 — 완전히 같은 기능
double = lambda x: x * 2
#         ↑        ↑   ↑
#       키워드  매개변수  반환식(return 생략)
```


````python
# 일반 함수 vs 람다 함수 비교
# ① 일반 함수
def calculate_bmi_normal(weight, height):
    return weight / (height ** 2)

# ② 람다 함수 (완전히 동일한 기능)
calculate_bmi_lambda = lambda weight, height: weight / (height ** 2)
# weight(체중), height(키)

# 사용법은 동일
bmi1 = calculate_bmi_normal(70, 1.75)
bmi2 = calculate_bmi_lambda(70, 1.75)

print(f"일반 함수 BMI: {bmi1:.2f}")
print(f"람다 함수 BMI: {bmi2:.2f}")
print(f"결과 동일: {bmi1 == bmi2}")
````

### 언제 람다, 언제 일반 함수? — 선택 기준

| 상황 | 선택 | 이유 |
|------|------|------|
| 한 줄로 표현 가능 | 람다 | 간결함 |
| 여러 줄 로직 필요 | 일반 함수 | 가독성 |
| 여러 번 재사용 | 일반 함수 | 이름 있어야 호출 가능 |
| sorted() key 등 임시 사용 | 람다 | 일회성 |
| 조건이 복잡 (3개 이상) | 일반 함수 | 람다는 읽기 어려워짐 |


````python
# ✅ 람다가 적합한 경우 — 단순, 일회성
good_lambda = lambda bp: "고혈압" if bp >= 140 else "정상"
patients_sorted = sorted(patients, key=lambda p: p["bp"])

# ❌ 람다가 부적합한 경우 — 복잡한 로직
# bad_lambda = lambda p: "위험" if (p["bp"] >= 160 and p["age"] >= 65) or ...

# ✅ 복잡한 로직은 일반 함수로
def assess_risk(patient):
    # assess(평가하다), risk(위험도)
    high_bp   = patient["bp"] >= 140
    elderly   = patient["age"] >= 65
    has_fever = patient["temp"] >= 37.5

    if high_bp and elderly:
        return "고위험"
    elif high_bp or has_fever:
        return "주의"
    else:
        return "안정"

for p in patients:
    risk = assess_risk(p)
    print(f"  {p['name']}: {risk}")
````

# 9. 타입 힌팅 (Type Hints)

타입 힌팅은 함수의 **입력과 출력 타입을 명시**해주는 방법입니다.
파이썬은 실행 시 강제하지 않지만, 코드 가독성과 협업에 크게 도움이 됩니다.

> 💡 **비유:** 택배 박스에 "깨지는 물건" 스티커를 붙이는 것.
> 규칙은 아니지만, 보는 사람이 어떻게 다뤄야 하는지 바로 알 수 있습니다.

```python
def 함수이름(매개변수: 타입) -> 반환타입:
```


````python
# 타입 힌팅 없는 함수 vs 있는 함수

# ① 힌팅 없음 — 매개변수에 무엇이 들어오는지 모름
def old_bmi(h, w):
    return w / (h / 100) ** 2

# ② 힌팅 있음 — 한눈에 입출력 타입을 알 수 있음
def calculate_bmi(height_cm: float, weight_kg: float) -> float:
    # height_cm: float(실수), weight_kg: float(실수)
    # 반환: float(실수)
    return round(weight_kg / (height_cm / 100) ** 2, 2)

# 반환값 없는 경우 -> None
def print_patient_info(name: str, age: int) -> None:
    # name: str(문자열), age: int(정수), 반환: None(없음)
    print(f"{name}({age}세)")

result = calculate_bmi(170.0, 75.0)
print(f"BMI: {result}")
print_patient_info("김환자", 45)
````

| | 나쁜 예 | 좋은 예 |
|-|--------|--------|

````python
# ❌ 나쁜 예: BMI 계산 + 출력 + 분류가 한 함수에 다 들어있음
def process_everything(name, height, weight):
    bmi = weight / (height / 100) ** 2      # 계산
    print(f"{name} BMI: {bmi:.2f}")         # 출력
    if bmi < 18.5: return "저체중"           # 분류
    elif bmi < 25: return "정상"
    else: return "과체중 이상"

# ✅ 좋은 예: 기능별로 함수 분리
def calculate_bmi(height_cm: float, weight_kg: float) -> float:
    # 계산만 담당
    return round(weight_kg / (height_cm / 100) ** 2, 2)

def classify_bmi(bmi: float) -> str:
    # 분류만 담당
    if bmi < 18.5: return "저체중"
    elif bmi < 25: return "정상"
    elif bmi < 30: return "과체중"
    else:          return "비만"

def print_bmi_report(name: str, height: float, weight: float) -> None:
    # 출력만 담당 — 위 두 함수를 조합
    bmi    = calculate_bmi(height, weight)    # 계산 함수 호출
    status = classify_bmi(bmi)               # 분류 함수 호출
    print(f"[{name}] BMI: {bmi:.2f} → {status}")

print_bmi_report("김환자", 170.0, 75.0)
print_bmi_report("이환자", 160.5, 58.2)
````

# 🏥 11. 미니 프로젝트 — AI 헬스케어 함수 라이브러리

> 🎯 **목표:** 1~4일차에서 배운 모든 개념을 **함수**로 포장해서
> 재사용 가능한 의료 데이터 처리 라이브러리를 만들어봅니다.

## 공통 데이터 — 이 셀을 먼저 실행하세요


````python
# 정상 기준값 (튜플 — 2일차, 3일차 연결)
BP_NORMAL   = (90, 139)    # BP_NORMAL(혈압 정상 범위)
TEMP_NORMAL = (36.1, 37.2) # TEMP_NORMAL(체온 정상 범위)
HR_NORMAL   = (60, 100)    # HR_NORMAL(심박수 정상 범위)
SPO2_NORMAL = (95, 100)    # SPO2_NORMAL(산소포화도 정상 범위)

# 병동 환자 데이터 (딕셔너리 리스트 — 2일차 연결)
ward_patients = [
    {"id":"P001","name":"김환자","age":72,
     "vitals":{"bp":158,"temp":38.4,"hr":108,"spo2":91},
     "allergies":{"페니실린","아스피린"}, "diagnosis":"고혈압,당뇨"},
    {"id":"P002","name":"이환자","age":45,
     "vitals":{"bp":122,"temp":36.8,"hr":75,"spo2":98},
     "allergies":{"설파제"}, "diagnosis":"편두통"},
    {"id":"P003","name":"박환자","age":68,
     "vitals":{"bp":165,"temp":37.0,"hr":88,"spo2":95},
     "allergies":{"이부프로펜"}, "diagnosis":"심부전"},
    {"id":"P004","name":"최환자","age":35,
     "vitals":{"bp":115,"temp":39.1,"hr":112,"spo2":93},
     "allergies":set(), "diagnosis":"폐렴"},
    {"id":"P005","name":"정환자","age":80,
     "vitals":{"bp":142,"temp":36.5,"hr":68,"spo2":97},
     "allergies":{"페니실린"}, "diagnosis":"만성 신부전"},
]

print(f"환자 데이터 로드 완료: {len(ward_patients)}명")
````

---
## ⭐ Step 1 — 활력징후 판정 함수
> 💡 **사용:** 비교 연산자(1일차) + 튜플 기준값(2일차) + if 조건문(3일차)

````python
def check_vital(value: float | int, normal_range: tuple, label: str) -> str:
    # value(측정값), normal_range(정상 범위 튜플), label(항목명)
    low, high = normal_range   # 튜플 언패킹
    if low <= value <= high:
        return f"✅ {label} {value} — 정상"
    else:
        direction = "높음" if value > high else "낮음"
        return f"⚠️ {label} {value} — {direction} (정상: {low}~{high})"

def assess_patient_vitals(patient: dict) -> list[str]:
    # assess(평가하다), vitals(활력징후)
    v       = patient["vitals"]
    results = []
    results.append(check_vital(v["bp"],   BP_NORMAL,   "혈압(mmHg)"))
    results.append(check_vital(v["temp"], TEMP_NORMAL, "체온(℃)"))
    results.append(check_vital(v["hr"],   HR_NORMAL,   "심박수(bpm)"))
    results.append(check_vital(v["spo2"], SPO2_NORMAL, "SpO2(%)"))
    return results

# 테스트
for patient in ward_patients[:2]:
    print(f"
[{patient['name']}]")
    for result in assess_patient_vitals(patient):
        print(f"  {result}")
````

---
## ⭐⭐ Step 2 — 위험도 분류 함수
> 💡 **사용:** 집합 연산(2일차) + 복합 조건(3일차) + return

````python
def classify_risk(patient: dict) -> str:
    # classify(분류하다), risk(위험도)
    v        = patient["vitals"]
    symptoms_like = {"고혈압", "심부전", "당뇨"}   # 고위험 진단명

    # 활력징후 이상 개수 계산 (for + if — 4일차 연결)
    abnormal = sum([
        not (BP_NORMAL[0]   <= v["bp"]   <= BP_NORMAL[1]),
        not (TEMP_NORMAL[0] <= v["temp"] <= TEMP_NORMAL[1]),
        not (HR_NORMAL[0]   <= v["hr"]   <= HR_NORMAL[1]),
        not (SPO2_NORMAL[0] <= v["spo2"] <= SPO2_NORMAL[1]),
    ])

    # 고위험 진단 여부 (집합 연산 — 2일차 연결)
    has_risky_diagnosis = any(
        s in patient["diagnosis"] for s in symptoms_like
    )

    # 조건 조합으로 위험도 결정
    if abnormal >= 3 or v["spo2"] < 90:
        return "🚨 응급"
    elif patient["age"] >= 65 and abnormal >= 2:
        return "⚠️ 고위험"
    elif has_risky_diagnosis and abnormal >= 1:
        return "⚠️ 주의"
    else:
        return "✅ 안정"

# 전체 환자 위험도 출력
print("=== 위험도 분류 결과 ===")
for p in ward_patients:
    risk = classify_risk(p)
    print(f"  {p['name']:5s}({p['age']}세): {risk}")
````

---
## ⭐⭐⭐ Step 3 — 처방 안전성 검사 함수
> 💡 **사용:** 집합 교집합(2일차) + return dict

````python
def check_prescription_safety(patient: dict, new_drugs: set) -> dict:
    # check(확인), prescription(처방), safety(안전성)
    conflict  = patient["allergies"] & new_drugs    # 교집합
    safe_rx   = sorted(new_drugs - patient["allergies"])  # 차집합

    return {
        "patient":  patient["name"],
        "safe":     len(conflict) == 0,
        "conflict": conflict,
        "safe_rx":  safe_rx,
        "message":  (f"⚠️ 충돌: {conflict}" if conflict
                     else "✅ 전체 처방 안전"),
    }

# 새 처방 약물
new_prescription = {"페니실린", "메트포르민", "암로디핀", "아스피린"}
# new_prescription(새 처방)

print("=== 처방 안전성 일괄 검사 ===")
for p in ward_patients:
    result = check_prescription_safety(p, new_prescription)
    print(f"  {result['patient']:5s}: {result['message']}")
    if not result["safe"]:
        print(f"         안전한 대체 처방: {result['safe_rx']}")
````

---
## ⭐⭐⭐⭐ Step 4 — 병동 통계 함수
> 💡 **사용:** 리스트 컴프리헨션(4일차) + sum/max/min/len(2일차) + return tuple

````python
def calculate_ward_stats(patients: list[dict]) -> tuple:
    # calculate(계산하다), ward(병동), stats(통계)
    bp_list   = [p["vitals"]["bp"]   for p in patients]
    temp_list = [p["vitals"]["temp"] for p in patients]
    age_list  = [p["age"]            for p in patients]

    fever_count  = sum(1 for p in patients if p["vitals"]["temp"] >= 37.5)
    high_bp_count= sum(1 for p in patients if p["vitals"]["bp"] >= 140)
    senior_count = sum(1 for p in patients if p["age"] >= 65)

    return (
        len(patients),
        round(sum(bp_list) / len(bp_list), 1),
        round(sum(temp_list) / len(temp_list), 1),
        round(sum(age_list) / len(age_list), 1),
        fever_count,
        high_bp_count,
        senior_count,
    )

# 튜플 언패킹으로 받기 (2일차 연결)
total, avg_bp, avg_temp, avg_age, fevers, high_bp, seniors =     calculate_ward_stats(ward_patients)

print("=== 병동 통계 대시보드 ===")
print(f"총 환자:      {total}명")
print(f"평균 혈압:    {avg_bp}mmHg")
print(f"평균 체온:    {avg_temp}℃")
print(f"평균 나이:    {avg_age}세")
print(f"발열 환자:    {fevers}명")
print(f"고혈압 환자:  {high_bp}명")
print(f"65세 이상:    {seniors}명")
````

---
## ⭐⭐⭐⭐⭐ Step 5 — 통합 환자 리포트 함수 (1~5일차 총동원)

````python
def generate_patient_report(patient: dict) -> str:
    # generate(생성하다), report(보고서)
    # ── 활력징후 판정 (Step 1 함수 재사용) ──────────────
    vital_results = assess_patient_vitals(patient)   # Step 1 함수 호출!
    vital_str     = "
".join(f"    {r}" for r in vital_results)

    # ── 위험도 (Step 2 함수 재사용) ──────────────────────
    risk = classify_risk(patient)                    # Step 2 함수 호출!

    # ── 활력징후 이상 항목만 필터 (람다 + 컴프리헨션) ────
    abnormal_items = [r for r in vital_results if "⚠️" in r]

    # ── f-string으로 리포트 조립 (1일차 연결) ─────────────
    sep  = "=" * 52
    dash = "-" * 52
    allergy_str = ", ".join(sorted(patient["allergies"])) or "없음"
    report = (
        f"
{sep}
"
        f"  환자 리포트  [{patient['id']}]
"
        f"{sep}
"
        f"  이름:   {patient['name']} ({patient['age']}세)
"
        f"  진단:   {patient['diagnosis']}
"
        f"  위험도: {risk}
"
        f"  알레르기: {allergy_str}
"
        f"{dash}
"
        f"  활력징후:
"
        f"{vital_str}
"
        f"{dash}
"
        f"  이상 항목: {len(abnormal_items)}개
"
        f"{sep}"
    )
    return report

# 전체 환자 리포트 출력
for p in ward_patients:
    print(generate_patient_report(p))
````

````python
# 도전 1: 환자를 위험도 기준으로 정렬하는 함수를 만들어보세요
# 힌트: sorted() + 람다 + classify_risk() 활용

# 도전 2: 특정 진단명의 환자만 필터링하는 함수를 만들어보세요
# def filter_by_diagnosis(patients, diagnosis_keyword): ...
# 힌트: 리스트 컴프리헨션 + in 연산자

# 도전 3: 처방 가능한 환자를 찾아서 리스트로 반환하는 함수를 만들어보세요
# def find_safe_patients(patients, new_drugs): ...
# 힌트: check_prescription_safety() 재사용

# 코드 작성
````

````python
# @title 정답 보기
# 도전 1: 위험도 기준 정렬
RISK_ORDER = {"🚨 응급": 0, "⚠️ 고위험": 1, "⚠️ 주의": 2, "✅ 안정": 3}
# RISK_ORDER(위험도 순서 딕셔너리)

def sort_by_risk(patients: list[dict]) -> list[dict]:
    # sort(정렬) + by_risk(위험도 기준)
    return sorted(patients,
                  key=lambda p: RISK_ORDER.get(classify_risk(p), 9))

print("=== 위험도 우선순위 정렬 ===")
for p in sort_by_risk(ward_patients):
    print(f"  {classify_risk(p)} | {p['name']} ({p['age']}세)")

print()

# 도전 2: 진단명 필터링
def filter_by_diagnosis(patients: list[dict], keyword: str) -> list[dict]:
    # keyword(검색어), filter(걸러내다)
    return [p for p in patients if keyword in p["diagnosis"]]

bp_patients = filter_by_diagnosis(ward_patients, "고혈압")
print(f"고혈압 환자: {[p['name'] for p in bp_patients]}")

print()

# 도전 3: 처방 안전 환자 찾기
def find_safe_patients(patients: list[dict], new_drugs: set) -> list[dict]:
    # find(찾다), safe(안전한)
    return [p for p in patients
            if check_prescription_safety(p, new_drugs)["safe"]]

test_rx = {"페니실린", "메트포르민"}
safe_pts = find_safe_patients(ward_patients, test_rx)
print(f"처방 안전 환자: {[p['name'] for p in safe_pts]}")
````

### 함수의 핵심 문법

| 개념 | 문법 | 핵심 |
|------|------|------|
| 함수 정의 | `def 이름(매개변수):` | `def` + 콜론 + 들여쓰기 |
| 함수 호출 | `이름(인수)` | 이름 + `()` |
| 매개변수 | 함수 정의 시 빈 자리 이름 | 받을 값의 이름표 |
| 인수 | 함수 호출 시 실제 값 | 실제로 전달하는 값 |
| 반환값 | `return 값` | 결과를 돌려줌 |
| 기본값 | `def f(x, y=10):` | 생략 가능한 인수 |
| 키워드 인수 | `f(y=5, x=3)` | 이름으로 전달 |
| 다중 반환 | `return a, b, c` | 튜플로 반환 |
| 람다 | `lambda x: x*2` | 한 줄 함수 |
| 타입 힌팅 | `def f(x: int) -> str:` | 타입 명시 |

### 1~5일차 연결 구조

```
1일차 변수·연산자     ──┐
2일차 자료구조         ──┼──► 5일차 함수로 모든 것을 묶어 재사용!
3일차 조건문           ──┤
4일차 반복문           ──┘
```
