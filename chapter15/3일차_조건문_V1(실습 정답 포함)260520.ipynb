| 분류 | 문법 | 예시 | 결과 |

# 🤔 1. 조건문이란?

조건문은 **주어진 조건에 따라 다른 코드를 실행**하는 구문입니다.
조건이 `True`면 실행하고, `False`면 건너뜁니다.

```
[조건 평가]
     │
  True?  ──── Yes ────► [코드 A 실행]
     │
    No
     │
     ▼
[코드 B 실행 (else)]
```

> 🏥 **의료 현장에서는?**
> 환자 체온이 37.5℃ 이상이면 → 해열제 처방
> 혈압이 140 이상이면 → 고혈압 약물 투여
> 알레르기 약물이 처방에 포함되면 → 처방 취소
>
> 이런 "만약 ~라면"의 판단을 코드로 표현하는 것이 조건문입니다.

## 파이썬 조건문의 핵심 규칙

```python
if 조건:           # ← 콜론(:) 필수
    실행할 코드    # ← 들여쓰기(4칸 또는 탭) 필수!
```

> ⚠️ **들여쓰기(indentation)가 매우 중요합니다.**
> 파이썬은 들여쓰기로 코드 블록을 구분합니다.
> 들여쓰기가 맞지 않으면 `IndentationError`가 발생합니다.

# 📋 2. if / if-else / if-elif-else

## 2-1. if — "만약 ~라면"


````python
# 가장 기본 형태 — 조건이 True일 때만 실행
temperature = 38.5   # temperature(체온)

if temperature >= 37.5:
    print("발열이 있습니다.")
    print("해열제를 처방합니다.")

print("체온 확인 완료")   # ← 들여쓰기 없음 = if 블록 밖 = 항상 실행
````

````python
# 조건이 False이면 블록 전체를 건너뜁니다
temperature = 36.5   # temperature(체온) — 정상 체온

if temperature >= 37.5:
    print("이 줄은 출력되지 않습니다")  # ← False이므로 건너뜀

print("항상 출력됩니다")
````

## 2-2. if-else — '아니라면'

````python
# 조건이 True일 때와 False일 때 각각 다른 코드 실행
patient_age = 16   # patient_age(환자 나이)

if patient_age >= 18:
    print("성인용 약물을 처방합니다.")
else:
    print("소아용 약물을 처방합니다.")

# else는 if 조건이 False일 때 무조건 실행됩니다
````

## 2-3. if-elif-else — '여러 조건 중 하나'

````python
# 여러 조건을 위에서부터 순서대로 확인 — 처음 True인 블록만 실행
bmi = 27.5   # bmi(body mass index, 체질량지수)

if bmi < 18.5:
    weight_status = "저체중"
elif bmi < 25.0:
    weight_status = "정상"
elif bmi < 30.0:
    weight_status = "과체중"
else:
    weight_status = "비만"

print(f"BMI {bmi} → {weight_status}")
````

````python
# elif 는 몇 개든 추가할 수 있습니다
# 수축기 혈압 단계 분류 (JNC 기준)
systolic = 145   # systolic(수축기의) 혈압

if systolic < 120:
    stage = "정상"
elif systolic < 130:
    stage = "높은 정상"
elif systolic < 140:
    stage = "고혈압 전단계"
elif systolic < 160:
    stage = "고혈압 1기"
else:
    stage = "고혈압 2기"

print(f"수축기 혈압 {systolic}mmHg → {stage}")
````

### ✏️ [직접해보기 2-1] 심박수 분류

````python
heart_rate = 52   # heart_rate(심박수), 단위: bpm

# 아래 기준으로 심박수를 분류하세요
# 60 미만    → "서맥 (Bradycardia)"
# 60 ~ 100  → "정상 (Normal)"
# 101 이상   → "빈맥 (Tachycardia)"
# 출력 예시: 심박수 52bpm → 서맥 (Bradycardia)

# 코드 작성
````

````python
# @title 정답 보기
heart_rate = 52

if heart_rate < 60:
    status = "서맥 (Bradycardia)"
elif heart_rate <= 100:
    status = "정상 (Normal)"
else:
    status = "빈맥 (Tachycardia)"

print(f"심박수 {heart_rate}bpm → {status}")
````

### ✏️ [직접해보기 2-2] 산소포화도 판정

````python
spo2 = 93   # spo2(oxygen saturation, 산소포화도), 단위: %

# 기준: 95 이상 → "정상"
#       90 ~ 94 → "경증 저산소증"
#       85 ~ 89 → "중등도 저산소증"
#       85 미만  → "중증 저산소증 — 즉시 산소 투여"

# 코드 작성
````

````python
# @title 정답 보기
spo2 = 93

if spo2 >= 95:
    status = "정상"
elif spo2 >= 90:
    status = "경증 저산소증"
elif spo2 >= 85:
    status = "중등도 저산소증"
else:
    status = "중증 저산소증 — 즉시 산소 투여"

print(f"산소포화도 {spo2}% → {status}")
````

# 🔬 3. 다양한 조건식 만들기

`if` 뒤에 오는 조건식은 **True 또는 False를 반환하는 모든 표현식**이 가능합니다.
1~2일차에서 배운 연산자와 자료구조를 조건식에 활용해봐요.

## 3-1. 비교 연산자 조건 (1일차 연결)


````python
# 비교 연산자 6가지를 조건식으로 활용
bp       = 145   # bp(blood pressure, 혈압)
age      = 70    # age(나이)
bmi      = 24.8  # bmi(체질량지수)
temp     = 36.8  # temp(temperature, 체온)
spo2     = 97    # spo2(산소포화도)

print("=== 비교 연산자 조건 ===")

if bp > 140:                      # > : 초과
    print(f"혈압 {bp} → 고혈압 의심")

if age >= 65:                     # >= : 이상
    print(f"나이 {age}세 → 노인 환자 주의")

if bmi < 25:                      # < : 미만
    print(f"BMI {bmi} → 정상 체중")

if temp <= 37.2:                  # <= : 이하
    print(f"체온 {temp}℃ → 정상")

if spo2 == 97:                    # == : 같다
    print(f"산소포화도 {spo2}% → 정확히 97%")

if bp != 120:                     # != : 다르다
    print(f"혈압 {bp} → 정상(120)과 다름")
````

## 3-2. 범위 조건 — 연속 비교 (파이썬 특유 문법)

````python
# 파이썬만의 특별한 문법: a <= x <= b (수학과 동일하게 표현)
bp   = 135   # bp(혈압)
temp = 37.0  # temp(체온)
spo2 = 96    # spo2(산소포화도)

# 튜플로 정상 기준값 저장 (2일차 연결)
BP_NORMAL   = (90, 139)     # BP_NORMAL(혈압 정상 범위)
TEMP_NORMAL = (36.1, 37.2)  # TEMP_NORMAL(체온 정상 범위)
SPO2_NORMAL = (95, 100)     # SPO2_NORMAL(산소포화도 정상 범위)

if BP_NORMAL[0] <= bp <= BP_NORMAL[1]:
    print(f"혈압 {bp}mmHg → 정상 범위 {BP_NORMAL}")

if TEMP_NORMAL[0] <= temp <= TEMP_NORMAL[1]:
    print(f"체온 {temp}℃ → 정상 범위 {TEMP_NORMAL}")

if not (SPO2_NORMAL[0] <= spo2 <= SPO2_NORMAL[1]):
    print(f"산소포화도 {spo2}% → 비정상!")
else:
    print(f"산소포화도 {spo2}% → 정상")
````

## 3-3. in / not in 조건 (2일차 리스트·집합 연결)

````python
# in 연산자: 리스트, 집합, 문자열에 값이 있는지 확인
symptoms      = ["발열", "기침", "두통", "근육통"]  # symptoms(증상들)
allergies     = {"페니실린", "아스피린", "설파제"}   # allergies(알레르기)
new_rx        = {"메트포르민", "아스피린", "암로디핀"} # new_rx(새 처방)
diagnosis_str = "고혈압성 심장병"                    # diagnosis_str(진단문자열)

print("=== in / not in 조건 ===")

# 리스트에서 증상 확인
if "발열" in symptoms:
    print("발열 감지 → 해열제 처방")

if "호흡곤란" not in symptoms:
    print("호흡곤란 없음 → 응급 조치 불필요")

# 집합 교집합으로 알레르기 충돌 확인 (2일차 연결)
conflict = allergies & new_rx   # conflict(충돌)
if conflict:                    # 교집합이 비어있지 않으면 Truthy
    print(f"알레르기 충돌! → {conflict}")

# 문자열 안에 특정 단어 포함 여부
if "고혈압" in diagnosis_str:
    print(f"진단명에 '고혈압' 포함 → 혈압약 처방 검토")
````

## 3-4. 논리 연산자 조건 — and / or / not (1일차 연결)

````python
# and: 모든 조건이 True여야 True
# or : 하나라도 True면 True
# not: True ↔ False 반전

age          = 72     # age(나이)
bp           = 155    # bp(혈압)
has_diabetes = True   # has_diabetes(당뇨 여부)
is_pregnant  = False  # is_pregnant(임신 여부)
has_allergy  = False  # has_allergy(알레르기 여부)

print("=== and 조건 ===")
# 모든 조건을 만족해야 실행
if age >= 65 and bp >= 140:
    print("고령 + 고혈압 → 집중 모니터링")

if age >= 18 and not has_allergy and not is_pregnant:
    print("성인 + 알레르기 없음 + 비임신 → 일반 약물 처방 가능")

print()
print("=== or 조건 ===")
# 하나라도 만족하면 실행
chest_pain   = False   # chest_pain(흉통)
spo2_low     = True    # spo2_low(산소포화도 저하)
unconscious  = False   # unconscious(의식 저하)

if chest_pain or spo2_low or unconscious:
    print("응급 증상 감지 → 즉시 처치 필요")

print()
print("=== not 조건 ===")
# 조건을 반전
if not is_pregnant:
    print("비임신 → NSAIDs 처방 가능")

has_taken_meds = False   # has_taken_meds(약 복용 여부)
if not has_taken_meds:
    print("미복용 확인 → 약 투여 진행")
````

## 3-5. Falsy 활용 조건 (1일차 연결)

````python
# Falsy: None / False / 0 / "" / [] / {} / set()
# if 조건에 값을 직접 넣으면 bool()이 자동 적용됩니다

# 빈 리스트 체크
medications = []   # medications(처방약들)
if not medications:
    print("처방약 없음 → 처방 필요")

medications.append("타이레놀")
if medications:    # 비어있지 않으면 True
    print(f"처방약 있음: {medications}")

# None 체크
patient_id = None   # patient_id(환자 ID)
if patient_id is None:
    print("환자 ID 없음 → 등록 필요")

patient_id = "P-2024-001"
if patient_id:      # None이 아닌 값은 Truthy
    print(f"환자 ID 확인: {patient_id}")

# 숫자 0 체크
pain_score = 0   # pain_score(통증 점수)
if not pain_score:
    print("통증 점수 0 → 통증 없음")

# 빈 문자열 체크
name = ""   # name(이름)
if not name:
    print("이름 미입력 → 입력 요청")

name = "  "   # 공백만 있는 경우
if not name.strip():   # strip()으로 공백 제거 후 확인
    print("이름이 공백만 있음 → 재입력 요청")
````

## 3-6. 딕셔너리·자료구조와 결합한 조건 (2일차 연결)

````python
# 2일차에서 만든 환자 차트 딕셔너리로 조건문 실습
patient = {
    "name":        "김환자",
    "age":         68,
    "diagnosis":   "고혈압, 당뇨",
    "allergies":   {"페니실린", "아스피린"},
    "medications": ["메트포르민", "암로디핀"],
    "vitals": {
        "bp":         158,
        "temp":       37.8,
        "heart_rate": 92,
        "spo2":       94
    }
}

print(f"=== {patient['name']} 상태 확인 ===")

# 딕셔너리 키로 접근한 값을 조건으로 사용
if patient["age"] >= 65:
    print(f"  고령 환자({patient['age']}세) → 낙상 위험 주의")

# 중첩 딕셔너리 접근
if patient["vitals"]["bp"] >= 140:
    print(f"  혈압 {patient['vitals']['bp']}mmHg → 고혈압 처치 필요")

if patient["vitals"]["temp"] >= 37.5:
    print(f"  체온 {patient['vitals']['temp']}℃ → 발열 처치")

# 진단명 문자열에 특정 질환 포함 여부
if "당뇨" in patient["diagnosis"]:
    print("  당뇨 진단 → 혈당 모니터링 필요")

# 집합으로 처방 안전성 검사
new_rx    = {"아스피린", "인슐린", "암로디핀"}   # new_rx(새 처방)
dangerous = patient["allergies"] & new_rx         # dangerous(위험 약물)
if dangerous:
    print(f"  ⚠️ 알레르기 충돌: {dangerous}")
````

## 3-7. isinstance() / type() 조건

````python
# isinstance(): 변수가 특정 타입인지 확인
# type(): 변수의 정확한 타입 반환

values = [42, 3.14, "홍길동", True, None, [1, 2], {"a": 1}]
# 다양한 타입의 값들

for v in values:
    if isinstance(v, int) and not isinstance(v, bool):
        print(f"{repr(v):15} → 정수(int)  — 나이, 혈압 등에 사용")
    elif isinstance(v, float):
        print(f"{repr(v):15} → 실수(float) — 체온, BMI 등에 사용")
    elif isinstance(v, str):
        print(f"{repr(v):15} → 문자열(str) — 이름, 진단명 등에 사용")
    elif isinstance(v, bool):
        print(f"{repr(v):15} → 불린(bool)  — 여부 표시에 사용")
    elif v is None:
        print(f"{repr(v):15} → None       — 값 없음")
    elif isinstance(v, list):
        print(f"{repr(v):15} → 리스트(list)")
    elif isinstance(v, dict):
        print(f"{repr(v):15} → 딕셔너리(dict)")
````

### ✏️ [직접해보기 3] 다양한 조건식 종합

# 🔍 4. 조건문 심화

## 4-1. 독립 조건 vs 다중 조건 vs 중첩 조건

| 방식 | 구조 | 특징 | 언제 쓰나? |
|------|------|------|-----------|
| 독립 조건 | `if / if / if` | 모든 조건을 각각 확인 | 여러 조건을 동시에 체크 |
| 다중 조건 | `if / elif / else` | 처음 True인 것만 실행 | 상호 배타적인 분류 |
| 중첩 조건 | `if` 안에 `if` | 단계별 확인 | 선행 조건이 있을 때 |


````python
patient = {
    "symptoms":  ["발열", "기침", "두통"],
    "age":       72,
    "bp":        165,
    "is_vip":    True
}

# ① 독립 조건 — 각각 별도로 확인 (모두 실행 가능)
print("=== 독립 조건 (증상별 처방) ===")
if "발열" in patient["symptoms"]:
    print("  해열제 처방")
if "기침" in patient["symptoms"]:
    print("  기침약 처방")
if "두통" in patient["symptoms"]:
    print("  진통제 처방")

print()

# ② 다중 조건 — 하나만 선택
print("=== 다중 조건 (나이별 병실 배정) ===")
if patient["age"] < 18:
    print("  소아 병동")
elif patient["age"] < 65:
    print("  일반 병동")
else:
    print("  노인 병동")   # ← age=72이므로 이것만 실행

print()

# ③ 중첩 조건 — 단계별 확인
print("=== 중첩 조건 (입원 처리 절차) ===")
if patient["bp"] >= 140:            # 1단계: 혈압 확인
    print("  혈압 이상 감지")
    if patient["age"] >= 65:        # 2단계: 나이 확인
        print("  고령 고혈압 → 즉시 입원")
        if patient["is_vip"]:       # 3단계: VIP 여부
            print("  VIP 병실 배정")
````

## 4-2. 삼항 연산자 — 한 줄 조건식

````python
# 문법: 참일때값 if 조건 else 거짓일때값
# 간단한 True/False 분기에 적합합니다

temp       = 38.2   # temp(체온)
bp         = 148    # bp(혈압)
age        = 70     # age(나이)
is_pregnant = False # is_pregnant(임신 여부)

# 기본 삼항 연산자
fever_status = "발열" if temp >= 37.5 else "정상"
print(f"체온 {temp}℃ → {fever_status}")

# 딕셔너리 접근과 결합
bp_status = "고혈압" if bp >= 140 else "정상"
print(f"혈압 {bp}mmHg → {bp_status}")

# not과 결합
rx_status = "처방 가능" if not is_pregnant else "처방 금기"
print(f"약물 처방 → {rx_status}")

# 중첩 삼항 (복잡해지면 if-elif-else가 더 나음)
age_group = "소아" if age < 18 else ("노인" if age >= 65 else "성인")
print(f"나이 {age}세 → {age_group}")
````

## 4-3. 조건식에서 자주 쓰는 문자열 메서드 (1일차 str 연결)

````python
# 조건식 안에 문자열 메서드를 바로 사용할 수 있습니다

diagnosis = "  고혈압성 심장병  "    # 앞뒤 공백 포함
drug_code = "ICD-10-I10"
blood_type = "a형"

# strip()으로 공백 제거 후 비교
if diagnosis.strip():               # 공백 제거 후 비어있지 않으면
    print(f"진단명: '{diagnosis.strip()}'")

# startswith()로 코드 형식 확인
if drug_code.startswith("ICD"):
    print(f"{drug_code} → ICD 코드 확인")

# upper()로 대소문자 통일 후 비교
if blood_type.upper() in ["A형", "B형", "O형", "AB형"]:
    print(f"혈액형 유효: {blood_type.upper()}")

# isdigit()으로 숫자 입력 검증
patient_id_input = "12345"   # 사용자 입력값
if patient_id_input.isdigit():
    patient_id = int(patient_id_input)   # 안전하게 변환
    print(f"환자 ID: {patient_id}")
````

### ✏️ [직접해보기 4] 삼항 + 중첩 조건 실습

````python
# 아래 값들로 삼항 연산자와 중첩 조건을 작성해보세요
spo2     = 92   # spo2(산소포화도)
has_mask = True # has_mask(마스크 착용 여부)

# 1. 삼항 연산자로 산소포화도 상태를 한 줄로 표현하세요
# spo2 >= 95이면 "정상", 아니면 "저산소증"
# spo2_status = ...

# 2. 중첩 조건으로 산소 처치 방법을 결정하세요
# 산소포화도 95 미만이면:
#   마스크 착용 중이면 → "고농도 산소 투여"
#   마스크 없으면     → "마스크 먼저 착용 후 산소 투여"
# 95 이상이면:
#   → "산소 처치 불필요"

# 코드 작성
````

````python
# @title 정답 보기
spo2     = 92
has_mask = True

# 1. 삼항 연산자
spo2_status = "정상" if spo2 >= 95 else "저산소증"
print(f"산소포화도 {spo2}% → {spo2_status}")

# 2. 중첩 조건
if spo2 < 95:
    if has_mask:
        print("고농도 산소 투여")
    else:
        print("마스크 먼저 착용 후 산소 투여")
else:
    print("산소 처치 불필요")
````

### ❓ 문제 4 — 처방 안전성 검사 (집합 교집합 조건, 2일차 연결)
> 💡 **사용 문법:** 집합 `&` 연산자, Falsy 활용 (`if conflict:`)


````python
patient_allergies = {"페니실린", "아스피린", "설파제"}  # 환자 알레르기
new_prescription  = {"메트포르민", "아스피린", "인슐린"}  # 새 처방

# 교집합으로 충돌 약물을 찾아 출력하세요
# 충돌 있으면: "⚠️ 알레르기 충돌: {충돌약물} — 처방 취소"
# 충돌 없으면: "✅ 처방 안전 확인 완료"

# 코드 작성
````

````python
# @title 정답 보기
patient_allergies = {"페니실린", "아스피린", "설파제"}
new_prescription  = {"메트포르민", "아스피린", "인슐린"}

conflict = patient_allergies & new_prescription   # conflict(충돌 약물)

if conflict:
    print(f"⚠️ 알레르기 충돌: {conflict} — 처방 취소")
else:
    print("✅ 처방 안전 확인 완료")
````

### ❓ 문제 5 — 혈압 5단계 분류 (elif 다중 조건)
> 💡 **사용 문법:** `if-elif-else` 다중 조건, 비교 연산자 체인


````python
systolic = 152   # systolic(수축기 혈압), mmHg

# JNC 기준 5단계 분류
# 120 미만         → "정상"
# 120 ~ 129        → "높은 정상"
# 130 ~ 139        → "고혈압 전단계"
# 140 ~ 159        → "고혈압 1기"
# 160 이상         → "고혈압 2기"
# 출력: "수축기 혈압 152mmHg → 고혈압 1기"

# 코드 작성
````

````python
# @title 정답 보기
systolic = 152

if systolic < 120:
    stage = "정상"
elif systolic < 130:
    stage = "높은 정상"
elif systolic < 140:
    stage = "고혈압 전단계"
elif systolic < 160:
    stage = "고혈압 1기"
else:
    stage = "고혈압 2기"

print(f"수축기 혈압 {systolic}mmHg → {stage}")
````

### ❓ 문제 6 — 리스트 슬라이싱 + 조건 (2일차 연결)
> 💡 **사용 문법:** 리스트 슬라이싱, `len()`, `max()`, `min()`, 비교 연산자


````python
bp_records = [120, 135, 148, 152, 145, 138, 142]
# bp_records(혈압 기록들) — 7일간 측정

# 최근 3일 기록만 슬라이싱
# 최근 3일 평균이 140 이상이면 "최근 혈압 악화 → 즉시 진료"
# 140 미만이면 "최근 혈압 안정"
# 전체 기록 중 최고값도 함께 출력

# 코드 작성
````

````python
# @title 정답 보기
bp_records = [120, 135, 148, 152, 145, 138, 142]

recent_3  = bp_records[-3:]                          # recent_3(최근 3일)
recent_avg = round(sum(recent_3) / len(recent_3), 1) # recent_avg(최근 평균)

print(f"전체 기록: {bp_records}")
print(f"최근 3일:  {recent_3}  (평균: {recent_avg}mmHg)")
print(f"전체 최고: {max(bp_records)}mmHg")

if recent_avg >= 140:
    print("최근 혈압 악화 → 즉시 진료")
else:
    print("최근 혈압 안정")
````

### ❓ 문제 7 — 딕셔너리 get() + 조건 (2일차 연결)
> 💡 **사용 문법:** `dict.get(key, 기본값)`, None 체크, Falsy 활용


````python
patient = {
    "name":      "박환자",
    "age":       55,
    "diagnosis": "당뇨",
    "allergies": {"설파제"},
}
# insurance_id 키가 없는 상태

# 1. get()으로 보험번호를 안전하게 조회하고,
#    없으면 "보험번호 미등록 → 자비 처리" 출력
# 2. 나이가 50 이상이면 "정기 검진 권장" 출력
# 3. 알레르기 집합이 비어있지 않으면 알레르기 목록 출력

# 코드 작성
````

````python
# @title 정답 보기
patient = {
    "name":      "박환자",
    "age":       55,
    "diagnosis": "당뇨",
    "allergies": {"설파제"},
}

# 1. get()으로 안전하게 조회
insurance = patient.get("insurance_id")   # insurance(보험번호)
if insurance is None:
    print("보험번호 미등록 → 자비 처리")
else:
    print(f"보험번호: {insurance}")

# 2. 나이 조건
if patient["age"] >= 50:
    print(f"{patient['name']}({patient['age']}세) → 정기 검진 권장")

# 3. 알레르기 집합 Falsy 활용
if patient["allergies"]:
    print(f"알레르기 약물: {patient['allergies']}")
````

````python
# 이 셀을 먼저 실행하세요 — 모든 Step에서 공통으로 사용합니다

# 정상 기준값 (튜플 — 2일차 연결)
BP_NORMAL    = (90,   139)    # BP_NORMAL(혈압 정상 범위)
TEMP_NORMAL  = (36.1, 37.2)   # TEMP_NORMAL(체온 정상 범위)
HR_NORMAL    = (60,   100)    # HR_NORMAL(심박수 정상 범위)
SPO2_NORMAL  = (95,   100)    # SPO2_NORMAL(산소포화도 정상 범위)

# 환자 차트 (중첩 딕셔너리 — 2일차 연결)
patient = {
    "id":        "P-2024-042",
    "name":      "홍환자",
    "age":       72,
    "gender":    "남",
    "diagnosis": "고혈압, 당뇨, 만성 신부전",
    "allergies": {"페니실린", "아스피린", "설파제"},   # 알레르기 집합
    "symptoms":  ["발열", "흉통", "호흡곤란", "부종"],  # 증상 리스트
    "vitals": {                                        # 활력징후 딕셔너리
        "bp":         162,
        "temp":       38.4,
        "heart_rate": 108,
        "spo2":       91,
    },
    "medications": ["암로디핀", "메트포르민"],           # 처방약 리스트
    "admitted":  False,                                # 입원 여부
}

print(f"환자 데이터 로드 완료: {patient['name']} ({patient['id']})")
print(f"나이: {patient['age']}세 | 진단: {patient['diagnosis']}")
````

## ⭐ Step 1 — 활력징후 정상 여부 판정

활력징후 4가지(혈압·체온·심박수·산소포화도)를 각각 정상 기준과 비교합니다.
> 💡 **사용 문법:** 비교 연산자, 튜플 인덱싱, 독립 if 조건


````python
# Step 1: 활력징후 정상 여부 판정
print(f"=== {patient['name']} 활력징후 판정 ===")
print(f"혈압: {patient['vitals']['bp']}mmHg | "
      f"체온: {patient['vitals']['temp']}℃ | "
      f"심박수: {patient['vitals']['heart_rate']}bpm | "
      f"산소포화도: {patient['vitals']['spo2']}%")
print()

# 코드 작성 — 각 활력징후가 정상 범위 밖이면 이상 메시지 출력
# 힌트: BP_NORMAL[0] <= bp <= BP_NORMAL[1] 패턴 사용
````

````python
# @title 정답 보기
print(f"=== {patient['name']} 활력징후 판정 ===")

bp   = patient["vitals"]["bp"]
temp = patient["vitals"]["temp"]
hr   = patient["vitals"]["heart_rate"]
spo2 = patient["vitals"]["spo2"]

if not (BP_NORMAL[0] <= bp <= BP_NORMAL[1]):
    print(f"  ⚠️ 혈압 {bp}mmHg → 정상 범위({BP_NORMAL[0]}~{BP_NORMAL[1]}) 벗어남")
else:
    print(f"  ✅ 혈압 {bp}mmHg → 정상")

if not (TEMP_NORMAL[0] <= temp <= TEMP_NORMAL[1]):
    print(f"  ⚠️ 체온 {temp}℃ → 정상 범위({TEMP_NORMAL[0]}~{TEMP_NORMAL[1]}) 벗어남")
else:
    print(f"  ✅ 체온 {temp}℃ → 정상")

if not (HR_NORMAL[0] <= hr <= HR_NORMAL[1]):
    print(f"  ⚠️ 심박수 {hr}bpm → 정상 범위({HR_NORMAL[0]}~{HR_NORMAL[1]}) 벗어남")
else:
    print(f"  ✅ 심박수 {hr}bpm → 정상")

if not (SPO2_NORMAL[0] <= spo2 <= SPO2_NORMAL[1]):
    print(f"  ⚠️ 산소포화도 {spo2}% → 정상 범위({SPO2_NORMAL[0]}~{SPO2_NORMAL[1]}) 벗어남")
else:
    print(f"  ✅ 산소포화도 {spo2}% → 정상")
````

## ⭐⭐ Step 2 — 응급 증상 감지 및 위험도 분류

증상 리스트와 활력징후를 조합해서 응급 여부와 위험도를 분류합니다.
> 💡 **사용 문법:** `in` 연산자(리스트), `or` 논리 연산자, `if-elif-else`


````python
# Step 2: 응급 증상 감지 및 위험도 분류
print(f"=== 응급 판정 ===")

# [도전 1] 아래 응급 증상 중 하나라도 있으면 "즉시 응급 처치 필요" 출력
# 응급 증상: "흉통", "호흡곤란", "의식저하"

# [도전 2] 활력징후 이상 개수를 세서 위험도 분류
# 이상 개수 3개 이상 → "위험"
# 이상 개수 1~2개   → "주의"
# 이상 없음         → "안정"
# 힌트: 변수 abnormal_count = 0 으로 시작해서 이상이면 += 1

# 코드 작성
````

````python
# @title 정답 보기
print(f"=== 응급 판정 ===")
symptoms = patient["symptoms"]
vitals   = patient["vitals"]

# 도전 1: 응급 증상 감지
emergency_symptoms = ["흉통", "호흡곤란", "의식저하"]   # emergency_symptoms(응급 증상들)
has_emergency = ("흉통" in symptoms or
                 "호흡곤란" in symptoms or
                 "의식저하" in symptoms)
# has_emergency(응급 여부)

if has_emergency:
    print("🚨 즉시 응급 처치 필요")
    if "흉통" in symptoms:
        print("  → 흉통 감지: 심전도 즉시 시행")
    if "호흡곤란" in symptoms:
        print("  → 호흡곤란: 산소 투여 시작")
else:
    print("응급 증상 없음")

print()

# 도전 2: 활력징후 이상 개수로 위험도 분류
abnormal_count = 0   # abnormal_count(이상 개수)

if not (BP_NORMAL[0]   <= vitals["bp"]         <= BP_NORMAL[1]):
    abnormal_count += 1
if not (TEMP_NORMAL[0] <= vitals["temp"]        <= TEMP_NORMAL[1]):
    abnormal_count += 1
if not (HR_NORMAL[0]   <= vitals["heart_rate"]  <= HR_NORMAL[1]):
    abnormal_count += 1
if not (SPO2_NORMAL[0] <= vitals["spo2"]        <= SPO2_NORMAL[1]):
    abnormal_count += 1

print(f"활력징후 이상 항목 수: {abnormal_count}개")

if abnormal_count >= 3:
    risk_level = "위험"
elif abnormal_count >= 1:
    risk_level = "주의"
else:
    risk_level = "안정"
# risk_level(위험도)

print(f"위험도: {risk_level}")
````

## ⭐⭐⭐ Step 3 — 처방 안전성 검사 + 약물 결정

알레르기, 임신 여부, 신장 기능 등을 종합해서 최종 처방을 결정합니다.
> 💡 **사용 문법:** 집합 `&` 교집합, `not`, 중첩 조건, `in` 연산자


````python
# Step 3: 처방 안전성 검사
print("=== 처방 안전성 검사 ===")

# 처방하려는 약물들
candidates = {
    "아스피린":    "항혈소판제",
    "페니실린":    "항생제",
    "메트포르민":  "당뇨약",
    "인슐린":      "당뇨약",
    "암로디핀":    "혈압약",
    "푸로세미드":  "이뇨제",
}
# candidates(처방 후보 약물들)

# [도전 1] 알레르기 충돌 약물 감지 후 안전한 약물만 추출
# [도전 2] 진단에 "당뇨"가 있으면 당뇨약 중 안전한 것 추천
# [도전 3] 진단에 "만성 신부전"이 있으면 "메트포르민" 사용 금기 추가 경고

# 코드 작성
````

````python
# @title 정답 보기
print("=== 처방 안전성 검사 ===")

candidates = {
    "아스피린":    "항혈소판제",
    "페니실린":    "항생제",
    "메트포르민":  "당뇨약",
    "인슐린":      "당뇨약",
    "암로디핀":    "혈압약",
    "푸로세미드":  "이뇨제",
}

candidate_set = set(candidates.keys())   # candidate_set(후보 집합)

# 도전 1: 알레르기 충돌 감지
conflict   = patient["allergies"] & candidate_set   # conflict(충돌)
safe_drugs = candidate_set - patient["allergies"]   # safe_drugs(안전 약물)

if conflict:
    print(f"⚠️ 알레르기 충돌 약물: {conflict}")
print(f"✅ 안전한 처방 후보: {safe_drugs}")

print()

# 도전 2: 진단에 "당뇨"가 있으면 당뇨약 추천
if "당뇨" in patient["diagnosis"]:
    print("당뇨 진단 → 당뇨약 검토")
    diabetes_drugs = {d for d, t in candidates.items() if t == "당뇨약"}
    # diabetes_drugs(당뇨약들)
    safe_diabetes  = diabetes_drugs & safe_drugs   # safe_diabetes(안전한 당뇨약)
    if safe_diabetes:
        print(f"  추천 당뇨약: {safe_diabetes}")

# 도전 3: 만성 신부전 + 메트포르민 금기
if "만성 신부전" in patient["diagnosis"]:
    print()
    print("⚠️ 만성 신부전 감지")
    if "메트포르민" in safe_drugs:
        print("  → 메트포르민: 신부전 환자 금기 — 인슐린으로 대체 권장")
````

## ⭐⭐⭐⭐ Step 4 — 입원 승인 시스템 (종합)

활력징후·응급 증상·나이·진단을 종합해서 입원 승인을 결정합니다.
> 💡 **사용 문법:** 중첩 조건, and/or 복합 조건, 삼항 연산자, 딕셔너리 수정


````python
# Step 4: 입원 승인 시스템
print("=== 입원 승인 판정 ===")

# [도전] 아래 규칙을 코드로 구현하세요
#
# 즉시 입원:
#   - 응급 증상(흉통/호흡곤란) 중 하나라도 있으면
#   - 또는 산소포화도 90% 미만이면
#
# 입원 권장:
#   - 65세 이상이고 활력징후 이상이 2개 이상이면
#   - 또는 체온 38.5 이상이면
#
# 외래 추적:
#   - 위 조건 미해당
#
# 최종 결정을 patient["admitted"] 에 저장 (True/False)
# 삼항 연산자로 admitted_str 도 만들어보세요

# 코드 작성
````

````python
# @title 정답 보기
print("=== 입원 승인 판정 ===")

vitals   = patient["vitals"]
symptoms = patient["symptoms"]

# 활력징후 이상 개수 계산
abnormal_count = 0
if not (BP_NORMAL[0]   <= vitals["bp"]         <= BP_NORMAL[1]): abnormal_count += 1
if not (TEMP_NORMAL[0] <= vitals["temp"]        <= TEMP_NORMAL[1]): abnormal_count += 1
if not (HR_NORMAL[0]   <= vitals["heart_rate"]  <= HR_NORMAL[1]): abnormal_count += 1
if not (SPO2_NORMAL[0] <= vitals["spo2"]        <= SPO2_NORMAL[1]): abnormal_count += 1

has_emergency = "흉통" in symptoms or "호흡곤란" in symptoms
# has_emergency(응급 여부)

# 입원 결정
if has_emergency or vitals["spo2"] < 90:
    decision = "즉시 입원"
    patient["admitted"] = True
    if has_emergency:
        print(f"  응급 증상 감지 → {decision}")
    if vitals["spo2"] < 90:
        print(f"  산소포화도 {vitals['spo2']}% 위험 → {decision}")

elif patient["age"] >= 65 and abnormal_count >= 2:
    decision = "입원 권장"
    patient["admitted"] = True
    print(f"  고령({patient['age']}세) + 활력징후 이상 {abnormal_count}개 → {decision}")

elif vitals["temp"] >= 38.5:
    decision = "입원 권장"
    patient["admitted"] = True
    print(f"  고열({vitals['temp']}℃) → {decision}")

else:
    decision = "외래 추적"
    patient["admitted"] = False
    print(f"  응급 아님 → {decision}")

# 삼항 연산자로 입원 상태 문자열
admitted_str = "입원" if patient["admitted"] else "외래"
# admitted_str(입원 상태 문자열)
print(f"
최종 결정: {decision} ({admitted_str})")
print(f"patient['admitted'] 업데이트: {patient['admitted']}")
````

````python
# Step 5: 최종 환자 리포트 생성
# 위 Step 1~4의 결과를 이 셀에서 종합 출력합니다
# (Step 1~4 셀을 먼저 실행한 후 이 셀을 실행하세요)

print("=" * 55)
print(f"  임상 판정 최종 리포트")
print("=" * 55)

# 기본 정보
print(f"환자번호:   {patient['id']}")
print(f"성  명:     {patient['name']}  ({patient['gender']}, {patient['age']}세)")
print(f"진  단:     {patient['diagnosis']}")
print("-" * 55)

# 활력징후 + 정상 여부 (삼항 연산자)
bp   = patient["vitals"]["bp"]
temp = patient["vitals"]["temp"]
hr   = patient["vitals"]["heart_rate"]
spo2 = patient["vitals"]["spo2"]

bp_ok   = "✅" if BP_NORMAL[0]   <= bp   <= BP_NORMAL[1]   else "⚠️"
temp_ok = "✅" if TEMP_NORMAL[0] <= temp <= TEMP_NORMAL[1] else "⚠️"
hr_ok   = "✅" if HR_NORMAL[0]   <= hr   <= HR_NORMAL[1]   else "⚠️"
spo2_ok = "✅" if SPO2_NORMAL[0] <= spo2 <= SPO2_NORMAL[1] else "⚠️"

print(f"혈    압:   {bp}mmHg {bp_ok}")
print(f"체    온:   {temp}℃ {temp_ok}")
print(f"심박수:     {hr}bpm {hr_ok}")
print(f"산소포화도: {spo2}% {spo2_ok}")
print("-" * 55)

# 증상
print(f"주요 증상:  {', '.join(patient['symptoms'])}")

# 알레르기
if patient["allergies"]:
    print(f"알레르기:   {', '.join(sorted(patient['allergies']))}")
else:
    print(f"알레르기:   없음")

# 처방약
if patient["medications"]:
    print(f"현재 처방:  {', '.join(patient['medications'])}")
print("-" * 55)

# 최종 판정
admitted_str = "입원" if patient["admitted"] else "외래"
print(f"입원 여부:  {admitted_str}")
print(f"입원 상태:  {patient['admitted']}")
print("=" * 55)
````

## 오늘 배운 것 정리

| 개념 | 문법 | 핵심 |
|------|------|------|
| 단일 조건 | `if 조건:` | 조건이 True일 때만 실행 |
| 양자 택일 | `if ... else:` | True/False 두 가지 경우 |
| 다중 분기 | `if ... elif ... else:` | 여러 조건 중 하나 선택 |
| 독립 조건 | `if / if / if` | 모든 조건을 각각 확인 |
| 중첩 조건 | `if` 안에 `if` | 단계별 조건 확인 |
| 삼항 연산자 | `A if 조건 else B` | 한 줄 조건식 |

## 조건식에 사용한 문법 (1~2일차 연결)

| 분류 | 문법 | 예시 |
|------|------|------|
| 비교 | `>` `<` `>=` `<=` `==` `!=` | `bp >= 140` |
| 범위 | `a <= x <= b` | `90 <= bp <= 139` |
| 포함 | `in` / `not in` | `"발열" in symptoms` |
| 논리 | `and` / `or` / `not` | `age >= 65 and bp >= 140` |
| 집합 | `&` 교집합 | `allergies & new_rx` |
| Falsy | `if 값:` | `if medications:` |
| 딕셔너리 | `dict["key"]` | `patient["age"]` |
| 문자열 | `.in`, `.strip()` | `"당뇨" in diagnosis` |
