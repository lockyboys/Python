#### [for문 작동 원리]

```

               인덱스 0     인덱스 1    인덱스 2

for  med  in  medications :
      ↑             ↑
  매 반복마다    이 리스트에서
  이 변수에      하나씩 꺼냄
  담김
```

#  2. enumerate() — 순서 번호와 값을 함께

## enumerate() 작동 원리

`enumerate()`는 리스트를 순회할 때 **번호와 값을 동시에** 가져옵니다.

```python
# 번호를 직접 세는 번거로운 방법
index = 0
for name in patients:
    print(index + 1, name)
    index += 1

# enumerate()로 간결하게
for num, name in enumerate(patients, 1):  # 1: 시작 번호
    print(num, name)
```

> 💡 `enumerate(리스트, 시작번호)` — 시작번호 생략 시 0부터 시작

#  6. for + if 결합 — 반복하며 조건 판단

반복문과 조건문을 결합하면 **"여러 데이터를 검사해서 조건에 맞는 것만 처리"** 할 수 있습니다.

## 핵심 패턴 3가지

```python
# 패턴 1 — 조건에 맞는 것만 출력
for x in 리스트:
    if 조건:
        print(x)

# 패턴 2 — 누적 카운트
count = 0
for x in 리스트:
    if 조건:
        count += 1

# 패턴 3 — 새 리스트에 모으기
result = []
for x in 리스트:
    if 조건:
        result.append(x)
```

> ⚠️ **들여쓰기 2단계!** for 안의 if는 8칸


### 예시 코드 1 — 조건에 맞는 것만 출력

## for vs while 선택 기준

| | for문 | while문 |
|-|-------|---------|
| 언제 | 반복 횟수가 **정해져 있을 때** | 조건이 만족될 때까지 |
| 예시 | 환자 명단 순회 | 혈압이 안정될 때까지 |
| 위험 | 없음 | 조건이 안 바뀌면 **무한 루프**! |

## while문 구조

```python
while 조건:
    실행할 코드
    조건을 바꾸는 코드  ← 반드시 필요!
```

> 영원히 멈추지 않습니다 (무한 루프).

```python
# 일반 for문으로 리스트 만들기
result = []
for 변수 in 리스트:
    result.append(변환식)

# 리스트 컴프리헨션 — 한 줄로!
result = [변환식 for 변수 in 리스트]

# 조건 필터 추가
result = [변환식 for 변수 in 리스트 if 조건]
```

# 🏥 11. 미니 프로젝트 — AI 헬스케어 병동 분석 시스템

> 🎯 **목표:** 오늘 배운 모든 반복문 개념으로 병동 데이터를 자동 분석합니다.
> 각 Step의 💡 힌트를 참고하여 순서대로 도전하세요!

## 공통 데이터 (먼저 실행)


````python
# 정상 기준값 (튜플)
BP_NORMAL    = (90, 139)
TEMP_NORMAL  = (36.1, 37.2)
HR_NORMAL    = (60, 100)
SPO2_NORMAL  = (95, 100)

# 병동 환자 데이터
ward_patients = [
    {"id":"P001","name":"김환자","age":72,
     "vitals":{"bp":158,"temp":38.4,"hr":108,"spo2":91},
     "allergies":{"페니실린","아스피린"}, "diagnosis":"고혈압,당뇨"},
    {"id":"P002","name":"이환자","age":45,
     "vitals":{"bp":122,"temp":36.8,"hr":75,"spo2":98},
     "allergies":{"설파제"}, "diagnosis":"편두통"},
    {"id":"P003","name":"박환자","age":68,
     "vitals":{"bp":165,"temp":37.0,"hr":88,"spo2":95},
     "allergies":{"이부프로펜"}, "diagnosis":"심부전"},
    {"id":"P004","name":"최환자","age":35,
     "vitals":{"bp":115,"temp":39.1,"hr":112,"spo2":93},
     "allergies":set(), "diagnosis":"폐렴"},
    {"id":"P005","name":"정환자","age":80,
     "vitals":{"bp":142,"temp":36.5,"hr":68,"spo2":97},
     "allergies":{"페니실린"}, "diagnosis":"만성 신부전"},
]
print(f"환자 데이터 로드 완료: {len(ward_patients)}명")
````

---
## ⭐ Step 1 — 전체 현황 출력
> 💡 `enumerate()` + f-string


````python
print("=" * 55)
print("  병동 환자 현황")
print("=" * 55)

for i, patient in enumerate(ward_patients, 1):
    v = patient["vitals"]
    print(f"{i}. [{patient['id']}] {patient['name']}({patient['age']}세)")
    print(f"   혈압:{v['bp']} | 체온:{v['temp']} | 심박:{v['hr']} | SpO2:{v['spo2']}%")
print()
print(f"총 환자: {len(ward_patients)}명")
````

---
## ⭐⭐ Step 2 — 활력징후 자동 판정
> 💡 `for` + `items()` + 튜플 언패킹 + `if`


````python
vital_info = {
    "bp":   {"label": "혈압",   "range": BP_NORMAL},
    "temp": {"label": "체온",   "range": TEMP_NORMAL},
    "hr":   {"label": "심박수", "range": HR_NORMAL},
    "spo2": {"label": "SpO2",   "range": SPO2_NORMAL},
}

print("=== 활력징후 자동 판정 ===")
for patient in ward_patients:
    v             = patient["vitals"]
    abnormal_list = []

    for key, info in vital_info.items():
        low, high = info["range"]
        value     = v[key]
        if not (low <= value <= high):
            abnormal_list.append(f"{info['label']}:{value}")

    status = "⚠️ 이상" if abnormal_list else "✅ 정상"
    print(f"  {patient['name']:5s}: {status}", end="")
    if abnormal_list:
        print(f"  [{', '.join(abnormal_list)}]")
    else:
        print()
````

---
## ⭐⭐⭐ Step 3 — 처방 안전성 검사
> 💡 `for` + 집합 교집합(2일차)


````python
new_prescription = {"페니실린", "메트포르민", "암로디핀", "아스피린"}

print("=== 처방 안전성 검사 ===")
print(f"처방 약물: {new_prescription}
")

for patient in ward_patients:
    conflict  = patient["allergies"] & new_prescription
    safe_rx   = sorted(new_prescription - patient["allergies"])

    if conflict:
        print(f"  {patient['name']:5s}: ⚠️ 충돌 {conflict}")
        print(f"         안전한 처방: {safe_rx}")
    else:
        print(f"  {patient['name']:5s}: ✅ 전체 처방 안전")
````

---
## ⭐⭐⭐⭐ Step 4 — 병동 통계 대시보드
> 💡 리스트 컴프리헨션 + `sum/max/min/len`


````python
# 컴프리헨션으로 각 수치 리스트 추출
bp_list   = [p["vitals"]["bp"]   for p in ward_patients]
temp_list = [p["vitals"]["temp"] for p in ward_patients]
age_list  = [p["age"]            for p in ward_patients]

# 통계 계산
fever_count  = sum(1 for p in ward_patients if p["vitals"]["temp"] >= 37.5)
high_bp_count= sum(1 for p in ward_patients if p["vitals"]["bp"] >= 140)
senior_count = sum(1 for p in ward_patients if p["age"] >= 65)

print("=== 병동 통계 ===")
print(f"총 환자:      {len(ward_patients)}명")
print(f"평균 혈압:    {sum(bp_list)/len(bp_list):.1f}mmHg")
print(f"평균 체온:    {sum(temp_list)/len(temp_list):.1f}℃")
print(f"평균 나이:    {sum(age_list)/len(age_list):.1f}세")
print(f"발열 환자:    {fever_count}명")
print(f"고혈압 환자:  {high_bp_count}명")
print(f"65세 이상:    {senior_count}명")
````
