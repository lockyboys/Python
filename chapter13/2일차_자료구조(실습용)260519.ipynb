# 2. 자료구조

자료구조는 **많은 양의 데이터를 체계적으로 다루는 방법**입니다.  
파이썬에는 **여러 데이터를 효율적으로 관리**하는 4가지 주요 자료구조가 있습니다.

| 자료구조 | 순서 | 변경 | 중복 | 표기법 | 활용 예시 |
| --- | --- | --- | --- | --- | --- |
| **리스트** | ⭕ | ⭕ | ⭕ | `[]` | 시간별 혈압기록, 처방순서 |
| **튜플** | ⭕ | ❌ | ⭕ | `()` | 환자 기본정보, 좌표값 |
| **딕셔너리** | 키-값 | ⭕ | 키 중복 ❌ | `{key: value}` | 환자정보, 진료기록 |
| **집합** | ❌ | ⭕ | ❌ | `{}` | 중복 처방되면 안되는 약물 |

#### 슬라이싱
여러 데이터를 범위로 가져오기

**슬라이싱 문법**

- `[시작:끝]`: 시작 인덱스부터 끝-1 인덱스까지
- `[:끝]`: 처음부터 끝-1 인덱스까지
- `[시작:]`: 시작부터 끝까지
- `[:]`: 전체
- `[시작:끝:간격]`: 시작부터 끝-1 인덱스까지 간격만큼 건너뛰기
- `[::간격]`: 처음부터 끝까지 간격만큼 건너뛰기

### 🛠️ 리스트 조작하기


#### ➕ 리스트에 요소 추가하기

| 메소드 | 설명 | 특징 |
| --- | --- | --- |
| `append()` | 리스트 끝에 **1개 요소** 추가 | 하나의 객체를 통째로 추가 |
| `insert(위치, 값)` | **원하는 위치**에 1개 요소 추가 | 특정 인덱스에 삽입 가능 |
| `extend()` | 리스트 끝에 **여러 요소** 한번에 추가 | 반복 가능한 객체의 각 요소를 개별 추가 |

````python
# 환자의 혈압 측정 기록 관리
blood_pressures = [120, 125, 130, 118]
blood_pressures
````

````python
# append: 1개 데이터를 맨 끝에 추가
blood_pressures.append(135)
print(f"새 측정 후: {blood_pressures}")
````

````python
# insert: 1개 데이터를 원하는 위치에 추가
blood_pressures.insert(1, 122)  # 인덱스 1 번째 위치에 추가
print(f"누락값 추가 후: {blood_pressures}")
````

````python
# extend: 여러 데이터를 맨 끝에 추가
evening_readings = [128, 122, 119]
blood_pressures.extend(evening_readings)
print(f"저녁 측정값 추가 후: {blood_pressures}")
````

````python
# append vs extend 차이점
test_readings = [140, 145]
blood_pressures.append(test_readings)  # 리스트가 하나의 요소로 추가됨
print(f"append로 추가: {blood_pressures}")
````

#### ➖ 리스트에서 요소 삭제하기

| 메소드 | 설명 | 반환값 | 특징 |
| --- | --- | --- | --- |
| `remove(값)` | **값**으로 삭제 | 없음 | 첫 번째 일치하는 것만 삭제 |
| `pop(인덱스)` | **인덱스**로 삭제 | 삭제된 값 반환 | 인덱스 생략 시 마지막 요소 |
| `clear()` | **모든 요소** 삭제 | 없음 | 리스트가 빈 상태가 됨 |
| `del` | **인덱스/슬라이스** 삭제 | 없음 | 변수 자체도 삭제 가능 |

````python
# 처방약 목록 관리
medications = ["아스피린", "메트포르민", "임시처방약", "리시노프릴", "아토르바스타틴"]
medications
````

````python
# remove: 값으로 삭제 (첫 번째 일치하는 것만 삭제)
medications.remove("임시처방약")
print(f"처방 제거 후: {medications}")
````

````python
# pop: 특정 인덱스로 삭제하고 삭제된 값 반환
removed_med = medications.pop(1)  # 인덱스 1번째 약물 삭제
print(f"삭제된 약물: {removed_med}")
print(f"처방 제거 후: {medications}")
````

````python
# 마지막 요소 삭제
last_med = medications.pop()
print(f"삭제된 약물: {last_med}")
print(f"처방 제거 후: {medications}")
````

````python
# clear: 리스트의 모든 요소 삭제 (모든 처방 중단)
medications.clear()
print(f"처방 제거 후: {medications}")
````

````python
# del: 변수 자체나 특정 인덱스 삭제
blood_pressures = [120, 125, 130, 118, 135, 128, 122]  # 혈압 측정값

del blood_pressures[0]    # 첫 번째 측정값 삭제
print(f"첫 측정값 제외: {blood_pressures}")

del blood_pressures[:2]   # 처음 2개 측정값 삭제
print(f"초기 측정값들 제외: {blood_pressures}")
````

#### 2. 정렬 메소드

리스트의 순서를 변경하는 메소드들입니다.

| 메소드 | 설명 | 원본 리스트 변경 여부 | 반환값 |
| --- | --- | --- | --- |
| `sorted()` | 새로운 정렬된 리스트 생성 | 변경 안됨 | 새로운 정렬된 리스트 |
| `sort()` | 원본 리스트 자체를 정렬 | 변경됨 | 없음 (None) |
| `reverse()` | 원본 리스트 순서를 뒤집기 | 변경됨 | 없음 (None) |

````python
# 환자별 체중 데이터 분석
patient_weights = [65.5, 78.2, 52.1, 89.7, 71.3]
print(f"원본 체중 데이터: {patient_weights}")
````

````python
# sorted(): 새로운 정렬된 리스트 반환 (원본 유지)
sorted_weights = sorted(patient_weights)
print(f"원본 유지: {patient_weights}")
print(f"정렬된 체중: {sorted_weights}")
````

````python
# sort(): 원본 리스트 자체를 정렬
patient_weights.sort()
print(f"원본 정렬됨: {patient_weights}")
````

````python
# reverse(): 리스트 순서를 뒤집기
patient_names = ["김철수", "이영희", "박민수", "최지연"]

patient_names.reverse()
print(f"뒤집어서 정렬: {patient_names}")
````

````python
# 특별한 정렬 방법들

departments = ['소아과', '내과', '산부인과', '외과', '정형외과'] # 진료과 목록

by_length = sorted(departments, key=len)  # 길이순 정렬
print(f"길이순: {by_length}")

# 가나다순 정렬
sorted_departments = sorted(departments)
print(f"가나다순: {sorted_departments}")

# 역순 정렬 (ㅎ부터 ㄱ까지)
reverse_departments = sorted(departments, reverse=True)
print(f"내림차순: {reverse_departments}")
````

#### 3. 변환 메소드

리스트를 다른 형태로 변환하는 메소드입니다.

````python
# 진료과 목록을 문자열로 변환
departments = ["내과", "외과", "소아과", "산부인과"]

# join(): 리스트를 문자열로 합치기
result1 = ", ".join(departments)
print(f"진료과 목록: {result1}")

result2 = " | ".join(departments)
print(f"진료과 구분: {result2}")

result3 = "".join(departments)
print(f"진료과 연결: {result3}")
````

## 📦 튜플 - 변경되지 않는 리스트

튜플은 리스트와 같이 여러 개의 데이터를 순서대로 저장하지만 한 번 만들면 변경할 수 없는 자료구조입니다.


### 튜플의 불변성 (Immutable)
#### 불변성이란?
불변성(immutable)이란 **객체가 생성된 후 그 내용을 변경할 수 없는 성질**을 말합니다.
```python
# 가변(mutable) vs 불변(immutable) 비교
patient_list = ["김환자", "65세"]    # 리스트: 가변
patient_tuple = ("김환자", "65세")   # 튜플: 불변

# 리스트는 내용 변경 가능
patient_list[1] = "66세"
print(f"리스트 변경 후: {patient_list}")

# 튜플은 내용 변경 불가능
patient_tuple[1] = "66세"  # 에러 발생!
```

튜플은 **불변성(immutable)**을 가진 자료구조입니다.
즉, 한 번 생성된 후에는 내용을 변경할 수 없습니다.


**왜 불변성이 중요한가?**
- **데이터 안전성**: 실수로 중요한 정보(환자 ID, 생년월일 등)가 변경되는 것을 방지
- **해시 가능**: 딕셔너리의 키로 사용 가능
- **메모리 효율성**: 같은 내용의 튜플은 메모리를 공유할 수 있음
```python
# 의료 정보에서 불변성 활용 예시
HOSPITAL_LOCATION = (37.5665, 126.9780)  # 병원 위치 좌표 (변경되면 안 됨)
EMERGENCY_CODE = ("119", "응급실", "24시간")  # 응급 연락처 (변경되면 안 됨)

print(f"병원 위치: {HOSPITAL_LOCATION}")
print(f"응급 연락처: {EMERGENCY_CODE}")
```

#### ➖ 딕셔너리 삭제
| 메소드 | 설명 | 반환값 | 특징 |
| --- | --- | --- | --- |
| `pop(키)` | **특정 키** 삭제 | 삭제된 값 반환 | 키가 없으면 기본값 반환 가능 |
| `popitem()` | **마지막 키-값 쌍** 삭제 | (키, 값) 튜플 반환 | 딕셔너리가 비어있으면 KeyError |
| `clear()` | **모든 키-값 쌍** 삭제 | 없음 | 빈 딕셔너리가 됨 |
| `del` | **특정 키** 또는 **변수 자체** 삭제 | 없음 | 슬라이싱 불가 |

````python
# 초기 환자 정보
patient = {
    "name": "김환자",
    "age": 65,
    "diagnosis": "고혈압",
    "phone": "010-1234-5678",
    "blood_type": "A형"
}
patient
````

````python
# pop(): 특정 키 삭제하고 값 반환
removed_phone = patient.pop("phone")
print(f"삭제된 전화번호: {removed_phone}")
print("전화번호 삭제 후:", patient)
````

````python
# pop() 기본값 설정 (키가 없을 때)
email = patient.pop("email", "이메일 정보 없음")
print(f"이메일: {email}")

print("현재 환자 정보:", patient)
````

````python
# popitem(): 마지막 키-값 쌍 삭제
last_item = patient.popitem() # 키-값 튜플로 반환
print(f"마지막 삭제 항목: {last_item}")
print("마지막 항목 삭제 후:", patient)
````

````python
# clear(): 모든 요소 삭제
patient.clear()
print("폐기 후 정보:", patient)
````

````python
# del: 특정 키나 변수 자체 삭제
patient = {"name": "박환자", "age": 45, "diagnosis": "감기"}

del patient["diagnosis"]  # 완치되어 진단명 삭제
print("삭제 후:", patient)
````

### 🔍 딕셔너리 안전한 접근
- get()메소드
- in 연산자

| 접근 방법 | 키가 없을 때 | 반환값 | 사용 시점 |
| --- | --- | --- | --- |
| `dict[키]` | KeyError 발생 | 해당 값 | 키가 확실히 존재할 때 |
| `dict.get(키)` | None 반환 | 해당 값 또는 None | 안전한 접근이 필요할 때 |
| `dict.get(키, 기본값)` | 기본값 반환 | 해당 값 또는 기본값 | 기본값을 설정하고 싶을 때 |
| `키 in dict` | False 반환 | True/False | 키 존재 여부만 확인할 때 |

### 🔄 집합 연산

집합 연산은 **두 집합 사이의 관계**를 계산하는 방법입니다.

- **연산자** (`|`, `&`, , `^`): 새로운 집합 반환
- **메소드** (`.union()`, `.intersection()` 등): 새로운 집합 반환
- **업데이트 메소드** (`.update()`, `.intersection_update()` 등): 원본 집합 변경

````python
# 두 환자의 복용 약물 비교
patient_a_drugs = {"아스피린", "메트포르민", "리시노프릴", "심바스타틴"}
patient_b_drugs = {"아토르바스타틴", "메트포르민", "암로디핀", "심바스타틴"}
````

#### 1) 합집합 (Union) - 두 집합의 모든 요소

````python
# 두 환자가 복용하는 모든 약물 종류

# 방법1: 연산자 활용
all_drugs_operator = patient_a_drugs | patient_b_drugs
print(f'전체 약물 (연산자): {all_drugs_operator}')

# 방법2: 메서드 활용
all_drugs_method = patient_a_drugs.union(patient_b_drugs)
print(f'전체 약물 (메서드): {all_drugs_method}')
````

#### 2) 교집합 (Intersection) - 두 집합의 공통 요소

````python
# 두 환자가 공통으로 복용하는 약물

# 방법1: 연산자 활용
common_drugs_operator = patient_a_drugs & patient_b_drugs
print(f'공통 복용약물 (연산자): {common_drugs_operator}')

# 방법2: 메서드 활용
common_drugs_method = patient_a_drugs.intersection(patient_b_drugs)
print(f'공통 복용약물 (메서드): {common_drugs_method}')
````

#### 3) 차집합 (Difference) - 첫 번째 집합에만 있는 요소

````python
# 환자 A만 복용하는 약물 (환자 B는 복용하지 않는 약물)

# 방법1: 연산자 활용
only_a_drugs_operator = patient_a_drugs - patient_b_drugs
print(f'환자 A 전용약물 (연산자): {only_a_drugs_operator}')

# 방법2: 메서드 활용
only_a_drugs_method = patient_a_drugs.difference(patient_b_drugs)
print(f'환자 A 전용약물 (메서드): {only_a_drugs_method}')
````

#### 4) 대칭차집합 (Symmetric Difference) - 둘 중 하나에만 있는 요소

````python
# 한 환자만 복용하는 약물들 (공통이 아닌 약물들)

# 방법1: 연산자 활용
unique_drugs_operator = patient_a_drugs ^ patient_b_drugs
print(f'각자만 복용하는 약물 (연산자): {unique_drugs_operator}')

# 방법2: 메서드 활용
unique_drugs_method = patient_a_drugs.symmetric_difference(patient_b_drugs)
print(f'각자만 복용하는 약물 (메서드): {unique_drugs_method}')
````

#### 원본 집합 업데이트 (In-place 연산)

원본 집합 자체를 변경하는 메소드들입니다.

````python
# 합집합 업데이트 - 환자 A의 처방에 환자 B의 약물 추가
patient_a_drugs = {"아스피린", "메트포르민", "리시노프릴", "심바스타틴"}
patient_b_drugs = {"아토르바스타틴", "메트포르민", "암로디핀", "심바스타틴"}

print(f'업데이트 전 환자 A: {patient_a_drugs}')
patient_a_drugs.update(patient_b_drugs)
print(f'합집합 업데이트 후: {patient_a_drugs}')
````

````python
# 교집합 업데이트 - 공통 약물만 남기기
patient_a_drugs = {"아스피린", "메트포르민", "리시노프릴", "심바스타틴"}
patient_b_drugs = {"아토르바스타틴", "메트포르민", "암로디핀", "심바스타틴"}

print(f'교집합 업데이트 전: {patient_a_drugs}')
patient_a_drugs.intersection_update(patient_b_drugs)
print(f'교집합 업데이트 후: {patient_a_drugs}')
````

````python
# 차집합 업데이트 - 환자 A만의 고유 약물만 남기기
patient_a_drugs = {"아스피린", "메트포르민", "리시노프릴", "심바스타틴"}
patient_b_drugs = {"아토르바스타틴", "메트포르민", "암로디핀", "심바스타틴"}

print(f'차집합 업데이트 전: {patient_a_drugs}')
patient_a_drugs.difference_update(patient_b_drugs)
print(f'차집합 업데이트 후: {patient_a_drugs}')
````

````python
# 대칭차집합 업데이트 - 공통이 아닌 약물들만 남기기
patient_a_drugs = {"아스피린", "메트포르민", "리시노프릴", "심바스타틴"}
patient_b_drugs = {"아토르바스타틴", "메트포르민", "암로디핀", "심바스타틴"}

print(f'대칭차집합 업데이트 전: {patient_a_drugs}')
patient_a_drugs.symmetric_difference_update(patient_b_drugs)
print(f'대칭차집합 업데이트 후: {patient_a_drugs}')
````

#### 4. all(), any() - 논리 검사


````python
# 모든 환자가 입원 중인지 확인
admission_status = [True, True, True, False]
all_admitted = all(admission_status)
any_admitted = any(admission_status)

print(f"모든 환자 입원 중: {all_admitted}")
print(f"입원 환자 존재: {any_admitted}")
````
