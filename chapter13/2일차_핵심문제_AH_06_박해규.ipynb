## ❓ 문제 1 — 리스트 인덱싱 · 슬라이싱

```python
blood_pressures = [120, 115, 130, 128, 135, 118, 122, 125]
#                    0    1    2    3    4    5    6    7
```

아래 결과가 출력되도록 인덱싱·슬라이싱을 작성하세요.

```
첫 번째: 120
마지막:  125
앞 3일:  [120, 115, 130]
최근 2일: [122, 125]
역순:    [125, 122, 118, 135, 128, 130, 115, 120]
```

````python
blood_pressures = [120, 115, 130, 128, 135, 118, 122, 125]

# 1. 첫 번째 값
print('첫 번째:', )
print(f"{blood_pressures[0]}")
# 2. 마지막 값 (음수 인덱스 사용)
print('마지막: ', )
print(f"{blood_pressures[-1]}")
# 3. 앞 3일
print('앞 3일: ', )
print(f"{blood_pressures[2]}")
# 4. 최근 2일
print('최근 2일:', )
print(f"{blood_pressures[-2:]}")
# 5. 전체 역순
print('역순:   ', )
print(f"{blood_pressures[::-1]}")
````

### 💡 힌트

| 문법 | 의미 |
|------|------|
| `[0]` | 첫 번째 |
| `[-1]` | 마지막 |
| `[:3]` | 처음부터 2번째까지 (3 미포함) |
| `[-2:]` | 뒤에서 2개 ~ 끝 |
| `[::-1]` | 전체 역순 |

## ❓ 문제 2 — 리스트 추가 · 삭제 메소드

```python
prescription = ['아스피린', '메트포르민', '인슐린']
```

순서대로 조작하고 각 단계마다 결과를 출력하세요.

```
1. append()  로 '암로디핀' 추가
2. insert()  로 맨 앞에 '클로피도그렐' 삽입
3. remove()  로 '인슐린' 삭제
4. 최종 리스트를 ' → ' 로 연결해 한 줄 출력
```

```
예상 최종 출력: 클로피도그렐 → 아스피린 → 메트포르민 → 암로디핀
```

````python
prescription = ['아스피린', '메트포르민', '인슐린']

# 1. append() — '암로디핀' 추가
prescription.append("암로디핀")
print('1:', prescription)

# 2. insert() — 맨 앞(인덱스 0)에 '클로피도그렐' 삽입
prescription.insert(0, "클로피도그렐")

print('2:', prescription)

# 3. remove() — '인슐린' 삭제
prescription.remove('인슐린')
print('3:', prescription)

# 4. ' → ' 로 연결해서 한 줄 출력
print('최종:','→'.join(prescription) )
````

### 💡 힌트

| 메소드 | 의미 |
|--------|------|
| `append(값)` | 맨 뒤에 추가 |
| `insert(위치, 값)` | 특정 위치에 삽입 |
| `remove(값)` | 해당 값 삭제 (첫 번째만) |
| `' → '.join(리스트)` | 리스트를 문자열로 연결 |

## ❓ 문제 3 — 튜플 생성 · 언패킹

아래 환자 검사 결과를 **하나의 튜플**로 만들고, 언패킹해서 출력하세요.

| 항목 | 값 |
|------|----|
| 이름 | `'이환자'` |
| 나이 | `72` |
| 진단명 | `'당뇨'` |
| 혈당 | `180` |

```
예상 출력: 이환자(72세) | 진단: 당뇨 | 혈당: 180mg/dL
```

````python
# 1. 튜플 생성
patient_info = ('이환자', 72, '당뇨', 180)

print(patient_info)
# 2. 언패킹
name, age, diagnosis, blood_sugar = patient_info

# 3. f-string으로 출력
print(f"{name}({age}세) | 진단: {diagnosis} | 혈당: {blood_sugar}mg/dL")
````

### 💡 힌트

```python
# 튜플은 () 로 만들고, 수정 불가
t = (1, 2, 3)

# 언패킹: 변수 개수 = 튜플 요소 개수
a, b, c = t   # a=1, b=2, c=3
```

> 튜플은 변경되면 안 되는 데이터(좌표, 기준값 등)에 사용합니다.

## ❓ 문제 4 — 딕셔너리 생성 · 조회 · 수정

아래 환자 정보로 딕셔너리를 만들고 요구사항을 수행하세요.

| 항목 | 값 |
|------|----|
| 이름 | `'최환자'` |
| 나이 | `60` |
| 진단 | `'당뇨'` |
| 담당의 | `'박의사'` |

```
1. 딕셔너리 생성
2. '전화번호': '010-9876-5432' 추가
3. get()으로 '보험번호' 조회 (없으면 '미등록' 출력)
4. f-string으로 아래 형식 출력
```

```
예상 출력:
[환자차트] 최환자(60세) | 진단: 당뇨 | 담당의: 박의사
전화번호: 010-9876-5432 | 보험번호: 미등록
```

````python
# 1. 딕셔너리 생성
patient = {
    "환자이름": '최환자',
    "환자나이": 60,
    "진단": '당뇨',
    "담당의": '박의사'
}
print(patient)
print(patient["환자이름"])
# 2. '전화번호' 추가
patient['전화번호'] = '010-9876-5432'
print(patient)

# 3. get()으로 '보험번호' 조회 (없으면 '미등록')
#insurance =
print(patient.get('보험번호', '미등록'))
# 4. f-string 출력
print(f"[{patient['환자이름']}차트] {patient['환자나이']}세 | 진단: {patient['진단']} | 담당의: {patient['담당의']}")
print(f"전화번호: {patient['전화번호']}")
````

### 💡 힌트

| 문법 | 의미 |
|------|------|
| `d['키'] = 값` | 추가 또는 수정 |
| `d['키']` | 조회 (없으면 KeyError) |
| `d.get('키', 기본값)` | 안전한 조회 (없으면 기본값 반환) |

## ❓ 문제 5 — 집합 연산

두 병동의 증상 데이터를 집합 연산으로 분석하세요.

```python
ward_a = {'발열', '기침', '두통', '오한'}
ward_b = {'발열', '구토', '두통', '설사'}
```

```
1. 두 병동 공통 증상      → 교집합
2. A 병동에만 있는 증상   → 차집합
3. 전체 증상 종류         → 합집합
```

```
예상 출력:
공통 증상: {'두통', '발열'}
A 병동만: {'기침', '오한'}
전체 증상: {'발열', '기침', '두통', '오한', '구토', '설사'}
```

````python
ward_a = {'발열', '기침', '두통', '오한'}
ward_b = {'발열', '구토', '두통', '설사'}

# 1. 교집합 — 두 병동 공통 증상
common = ward_a & ward_b
print('공통 증상:', common)

# 2. 차집합 — A 병동에만 있는 증상
only_a = ward_a - ward_b
print('A 병동만:', only_a)

# 3. 합집합 — 전체 증상 종류
all_symptoms = ward_a | ward_b
print('전체 증상:', all_symptoms)
````

### 💡 힌트

| 연산자 | 의미 |
|--------|------|
| `A & B` | 교집합 — 공통 요소 |
| `A - B` | 차집합 — A에만 있는 것 |
| `A \| B` | 합집합 — 전체 |

> 집합은 **중복 없음, 순서 없음** — 중복 제거가 필요할 때 유용합니다.

## ❓ 문제 6 — 4종 자료구조 연계 실습 🚀

아래 데이터를 활용해서 환자 차트를 완성하세요.

```python
patient_name = '홍길동'
temp_records = [36.5, 36.8, 37.1, 38.2, 37.5, 36.9]  # 체온 기록
allergy_set  = {'페니실린', '이부프로펜'}               # 알레르기
new_rx_drugs = {'타이레놀', '이부프로펜', '메트포르민'} # 신규 처방
```

```
1. 체온 평균·최고·최저를 튜플로 만드세요
2. 알레르기와 처방 약물의 충돌 약물을 집합 연산으로 찾으세요
3. 안전한 처방 = 신규 처방 - 충돌 약물
4. 결과를 f-string으로 출력
```

```
예상 출력:
[홍길동 차트]
체온 기록: [36.5, 36.8, 37.1, 38.2, 37.5, 36.9]
체온 통계: 평균 37.2℃ | 최고 38.2℃ | 최저 36.5℃
⚠️ 충돌 약물: {'이부프로펜'}
✅ 안전 처방: ['메트포르민', '타이레놀']
```

````python
patient_name = '홍길동'
temp_records = [36.5, 36.8, 37.1, 38.2, 37.5, 36.9]
allergy_set  = {'페니실린', '이부프로펜'}
new_rx_drugs = {'타이레놀', '이부프로펜', '메트포르민'}

# 1. 체온 통계 튜플 (평균은 소수점 1자리)
temp_stats = tuple(temp_records)
avg = round(sum(temp_stats) / len(temp_stats), 1)
high = max(temp_stats)
low = min(temp_stats)

print(f"체온 통계: 평균 {avg}℃ | 최고 {high}℃ | 최저 {low}℃")


# 2. 충돌 약물 (교집합)
conflict = allergy_set & new_rx_drugs

# 3. 안전 처방 (차집합 → 정렬된 리스트)
safe_rx = allergy_set - new_rx_drugs
safe_rx_set = list(safe_rx)

safe_rx_set.sort()
print(f"차집합 → 정렬된 리스트 : {safe_rx_set}")
# 4. 출력
print(f'[{patient_name} 차트]')
print(f'체온 기록: {temp_records}')
print(f'체온 통계: 평균 {avg}℃ | 최고 {high}℃ | 최저 {low}℃')
print(f'⚠️ 충돌 약물: {conflict}')
print(f'✅ 안전 처방: {safe_rx}')
````

### 💡 힌트

| 함수 | 의미 |
|------|------|
| `sum(리스트)` | 합계 |
| `len(리스트)` | 개수 |
| `max(리스트)` | 최댓값 |
| `min(리스트)` | 최솟값 |
| `round(값, 자리수)` | 반올림 |
| `sorted(집합/리스트)` | 정렬된 리스트 반환 |
