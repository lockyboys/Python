### 데모 ① 계산기 대신 파이썬으로 계산하기

````python
# 지금은 이해 안 해도 됩니다. 그냥 실행해보세요!

# 환자의 혈압 정보
systolic = 145   # systolic(수축기의) 혈압
normal  = 120    # normal(정상) 기준값

# 정상 기준과 얼마나 차이나는지 자동 계산
diff = systolic - normal   # diff = difference(차이)

print(f"환자 혈압: {systolic}mmHg")
print(f"정상 기준: {normal}mmHg")
print(f"정상보다 {diff}mmHg 높습니다.")
````

### 데모 ② 환자 안내 문자 자동으로 만들기

````python
# 지금은 이해 안 해도 됩니다. 그냥 실행해보세요!

# 환자 정보를 입력하면
patient_name = "김환자"   # patient(환자) + name(이름)
room_number  = 304        # room(병실) + number(번호)
doctor_name  = "이의사"   # doctor(의사) + name(이름)

# 안내 문자가 자동으로 완성됩니다
message = f"""{patient_name} 환자분께 안내드립니다.
담당 의사: {doctor_name} 선생님
병실 번호: {room_number}호
내일 오전 9시에 진료가 예정되어 있습니다."""

print(message)
````

### 데모 ③ 조건 자동 판정하기

````python
# 지금은 이해 안 해도 됩니다. 그냥 실행해보세요!

# 체온을 입력하면
body_temp = 38.5   # body(신체) + temp = temperature(체온)

# 발열 여부를 자동으로 판정합니다
if body_temp >= 37.5:
    print(f"체온 {body_temp}℃ → ⚠️  발열입니다. 해열제를 처방합니다.")
else:
    print(f"체온 {body_temp}℃ → ✅ 정상 체온입니다.")
````

# 1. 출력부터 해봐요 — print()

코딩을 배울 때 가장 먼저 배우는 것이 **화면에 내용을 출력하는 것**입니다.
결과를 눈으로 확인해야 내가 코드를 올바르게 작성했는지 알 수 있거든요!

파이썬에서는 `print()` 함수를 사용합니다.

> 💡 **함수(function)란?** 특정 기능을 수행하는 명령어 묶음입니다.
> `print()`는 괄호 안의 내용을 화면에 출력해주는 함수예요.


---

## 함수(Function)란? — 내장 함수 vs 외장 함수

`print()`는 **함수(function)** 의 일종입니다.
함수란 특정 기능을 수행하는 명령어 묶음으로, `함수이름()` 형태로 호출합니다.

파이썬의 함수는 크게 두 종류로 나뉩니다.

| 구분 | 내장 함수 (Built-in Function) | 외장 함수 (External Library) |
|------|-------------------------------|------------------------------|
| 뜻 | 파이썬 설치 시 기본 제공 | 별도 설치·불러오기 필요 |
| 사용법 | 바로 사용 가능 | `import 라이브러리명` 후 사용 |
| 예시 | `print()`, `len()`, `type()` | `pandas`, `numpy`, `requests` |

> 💡 **비유로 이해하기**
> - 내장 함수 = 스마트폰에 기본 탑재된 앱 (전화, 문자, 카메라)
> - 외장 함수 = 필요할 때 앱스토어에서 설치하는 앱 (카카오톡, 유튜브)


### 자주 쓰이는 내장 함수 (Built-in Functions)

내장 함수는 `import` 없이 **바로 사용**할 수 있습니다.
아래 함수들은 실무에서도 매우 자주 쓰이니 익혀두면 유용합니다!


````python
# ① print() — 화면 출력
# print(출력할 값)
print("안녕하세요!")
print(42, "세", 170.5, "cm")
````

````python
# ② type() — 데이터 타입 확인
# type(변수 또는 값)
print(type("홍길동"))
print(type(25))
print(type(36.5))
print(type(True))
````

````python
# ③ len() — 길이 확인 (length, 길이)
# 문자열의 글자 수, 리스트의 항목 수 등을 셉니다
patient_name = "김환자"
diagnosis    = "고혈압성 심장질환"   # diagnosis(진단명)

print(len(patient_name))
print(len(diagnosis))
print(len("Hello"))
````

````python
# ④ int() / float() / str() — 타입 변환
# 데이터 타입을 변환합니다 (5. 데이터 타입 섹션에서 자세히 다룹니다)
print(int("42"))
print(float("36.5"))
print(str(100))
````

````python
# ⑤ round() — 반올림 (round, 반올림하다)
# round(값, 소수점 자리수)
bmi = 23.456789   # bmi = body mass index(체질량지수)

print(round(bmi))
print(round(bmi, 1))
print(round(bmi, 2))
````

````python
# ⑥ abs() — 절댓값 (absolute value, 절댓값)
# 음수를 양수로 바꿔줍니다
normal_bp = 120        # normal(정상) + bp(blood pressure, 혈압)
patient_bp = 95        # patient(환자) + bp(혈압)

difference = patient_bp - normal_bp
print(f"차이: {difference}")
print(f"절댓값: {abs(difference)}")
````

````python
# ⑦ max() / min() — 최댓값 / 최솟값
# max = maximum(최대), min = minimum(최소)
# 여러 값 중 가장 크거나 작은 값을 찾습니다

# 환자 3명의 혈당 수치
sugar_a = 95
sugar_b = 134
sugar_c = 112

print(f"가장 높은 혈당: {max(sugar_a, sugar_b, sugar_c)}")
print(f"가장 낮은 혈당: {min(sugar_a, sugar_b, sugar_c)}")
````

````python
# ⑧ input() — 사용자 입력 받기
# 프로그램 실행 중에 사용자가 값을 직접 입력하게 합니다
# (아래 셀을 실행하면 입력창이 나타납니다 — 아무 값이나 입력해보세요!)

name = input("이름을 입력하세요: ")   # input(입력)
print(f"안녕하세요, {name}님!")

# ⚠️ input()의 결과는 항상 문자열(str)입니다
# 숫자로 사용하려면 int() 또는 float()로 변환해야 합니다
````

외장 함수는 필요할 때 `import 라이브러리명` 으로 불러와서 사용합니다.

> 💡 **라이브러리(library)** 란? 다른 사람이 미리 만들어둔 함수들의 묶음입니다.
> 바퀴를 다시 발명할 필요 없이 가져다 쓰면 됩니다!


````python
# 외장 라이브러리는 import 로 불러옵니다
# import 라이브러리이름

# 예시: math 라이브러리 — 수학 계산
import math   # math(수학) 라이브러리

print(math.sqrt(144))
print(math.pi)
print(math.ceil(3.2))
print(math.floor(3.8))
````

## f-string — 변수를 문장 안에 넣는 방법 (강력 추천!)

변수를 배웠으니 이제 변수를 출력에 활용하는 더 좋은 방법을 배워볼게요.  
문자열을 출력하는데 있어 다양 방법이 있는데 이를 포메팅이라 합니다.

앞에서 배운 쉼표 방법과 f-string을 비교해보세요.

```python
# 쉼표 방법 (기본)
print("이름:", name, "나이:", age, "살")

# f-string 방법 (더 깔끔!)
print(f"이름: {name} 나이: {age}살")
```

`f"..."` 형태로 문자열을 쓰고, 변수를 넣고 싶은 자리에 `{ }` 를 씁니다.
`f`는 **format(형식을 만들다)** 의 줄임말입니다.


````python
# f-string 기본 사용법
# 따옴표 바로 앞에 f 를 붙이고, 변수 자리에 { 변수명 } 을 씁니다

name   = "이민수"   # name(이름)
age    = 28         # age(나이)
height = 180.5      # height(키)
````

규칙을 어기면 파이썬이 오류를 발생시킵니다.

## 타입 변환 (Type Casting)

타입 변환이란 데이터 타입을 다른 타입으로 바꾸는 것입니다.

> 💡 **왜 필요할까요?**
> 프로그램에서 사용자 입력이나 파일 데이터는 대부분 **문자열(str)** 로 들어옵니다.
> 계산을 하려면 숫자(int 또는 float)로 변환해야 합니다.


````python
# 타입 변환 함수
# int()   : 정수로 변환 (integer)
# float() : 실수로 변환 (floating point)
# str()   : 문자열로 변환 (string)

age_input    = "25"     # 문자열 "25" (따옴표 있음)
height_input = "172.4"  # 문자열 "172.4"

# 문자열 → 숫자로 변환
age    = int(age_input)       # "25" → 25
height = float(height_input)  # "172.4" → 172.4

print(f"변환 전: '{age_input}' (타입: {type(age_input)})")
print(f"변환 후: {age}  (타입: {type(age)})")

# 숫자 → 문자열로 변환
age_str = str(age)   # 25 → "25"
print(f"다시 문자열로: '{age_str}' (타입: {type(age_str)})")
````

### 🏥 의료 활용 예시 ① — 약물 용량 계산

````python
# 약물 용량 계산
weight      = 70   # weight(체중), 단위: kg
dose_per_kg = 5    # dose(용량) + per(~당) + kg = kg당 용량, 단위: mg
times_a_day = 3    # times(횟수) + a_day(하루에) = 하루 복용 횟수

total_dose  = weight * dose_per_kg         # total(합계) + dose(용량) = 총 투여량
single_dose = total_dose / times_a_day     # single(1회) + dose(용량) = 1회 용량

print(f"총 투여량: {total_dose}mg")
print(f"1회 용량: {single_dose:.1f}mg (하루 {times_a_day}회)")
````

### 🏥 의료 활용 예시 ② — 처방 약 포장 계산 (`//` 와 `%` 활용)

````python
# 처방 약 포장 계산
total_pills     = 28    # total(총) + pills(알약 수) = 총 알약 수
pills_per_pack  = 7     # pills(알약) + per(~당) + pack(포장) = 포장 당 알약 수

# // : 몫 — 완성된 포장 수
full_packs = total_pills // pills_per_pack   # full(완전한) + packs(포장 수)

# % : 나머지 — 남는 알약 수
remaining  = total_pills % pills_per_pack    # remaining(남은)

print(f"총 알약: {total_pills}개")
print(f"포장 단위: {pills_per_pack}개")
print(f"완성 포장: {full_packs}봉지")
print(f"남는 알약: {remaining}개")
print()
print(f"검산: {pills_per_pack} × {full_packs} + {remaining} = {pills_per_pack * full_packs + remaining}")
````

## 복합 대입 연산자 (Compound Assignment Operators)

복합(compound) 대입 연산자는 **계산과 저장을 한 번에** 처리합니다.

> 💡 **왜 필요할까요?**
> `stock = stock + 50` 처럼 같은 변수명을 두 번 쓰는 것을 줄여줍니다.
> `stock += 50` 으로 더 짧고 읽기 쉽게 쓸 수 있습니다.

## 비교 연산자 (Comparison Operators)

비교(comparison) 연산자는 두 값을 비교해서 **True(참) 또는 False(거짓)** 를 반환합니다.
결과가 항상 불린(bool) 타입이라는 게 핵심입니다!


````python
# 비교 연산자 6가지
x = 10
y = 5

print(f"{x} >  {y} = {x > y}")   # 크다 (greater than)
print(f"{x} <  {y} = {x < y}")   # 작다 (less than)
print(f"{x} >= {y} = {x >= y}")  # 크거나 같다 (greater than or equal to)
print(f"{x} <= {y} = {x <= y}")  # 작거나 같다 (less than or equal to)
print(f"{x} == {y} = {x == y}")  # 같다 (equal) — = 두 개! (= 하나는 대입)
print(f"{x} != {y} = {x != y}")  # 같지 않다 (not equal)
````

## 논리 연산자 (Logical Operators)

논리(logical) 연산자는 여러 조건을 **조합**할 때 사용합니다.
결과는 항상 True 또는 False입니다.

> 📋 **논리 연산자 진리표**
>
> | A | B | A and B | A or B | not A |
> |---|---|---------|--------|-------|
> | True | True | **True** | True | False |
> | True | False | **False** | True | False |
> | False | True | **False** | True | True |
> | False | False | **False** | **False** | True |
>
> - `and` : **둘 다** True여야 True
> - `or`  : **하나라도** True면 True
> - `not` : True ↔ False 반전


### 🤝 and 연산자 — 둘 다 True여야 True

````python
# and: 두 조건이 모두 True일 때만 True

# 예시 1: 코로나 의심 조건
has_fever = True    # has(있다) + fever(열) = 발열 여부
has_cough = False   # has(있다) + cough(기침) = 기침 여부

# 발열 AND 기침 — 둘 다 있어야 의심
covid_suspect = has_fever and has_cough   # covid(코로나) + suspect(의심)
print(f"발열: {has_fever}, 기침: {has_cough}")
print(f"코로나 의심: {covid_suspect}")   # False (기침이 없으므로)

print()

# 예시 2: 고위험군 조건
age       = 70   # age(나이)
has_fever = True # has(있다) + fever(열) = 발열 여부

# 65세 이상 AND 발열
high_risk = (age >= 65) and has_fever   # high(높은) + risk(위험) = 고위험
print(f"나이: {age}세, 발열: {has_fever}")
print(f"고위험군: {high_risk}")   # True (둘 다 True)
````
